## 📚 El concepto: valor posicional

En base 10 cada cifra vale según su **posición**: unidades, decenas, centenas… cada posición vale **10 veces** más que la anterior.

$$2026 = 2 \cdot 1000 + 0 \cdot 100 + 2 \cdot 10 + 6 \cdot 1 = 2 \cdot 10^3 + 0 \cdot 10^2 + 2 \cdot 10^1 + 6 \cdot 10^0$$

En **base 2** (binario) solo hay dos cifras, **0 y 1**, y cada posición vale **el doble** que la anterior: 1, 2, 4, 8, 16, 32, …

$$1101_2 = 1 \cdot 8 + 1 \cdot 4 + 0 \cdot 2 + 1 \cdot 1 = 13$$

Las computadoras usan binario porque un circuito tiene dos estados fáciles de distinguir: **apagado (0)** y **encendido (1)**. Cada 0 o 1 se llama **bit**.

## 1. Del binario al decimal

```python
def binario_a_decimal(texto):
    """Convierte una cadena como '1101' a su valor decimal."""
    valor = 0
    potencia = 1                        # la posición de la derecha vale 1
    for cifra in reversed(texto):       # recorremos de derecha a izquierda
        if cifra == "1":
            valor = valor + potencia
        potencia = potencia * 2         # cada posición vale el doble
    return valor

for b in ["1", "10", "101", "1101", "11111111", "10000000000"]:
    print(b, "→", binario_a_decimal(b))
```

```python
# Python ya sabe hacerlo: int(texto, base)
print(int("1101", 2))
print(int("11111111", 2))
```

## 2. Del decimal al binario: dividir entre 2

Para pasar 13 a binario dividimos entre 2 una y otra vez y anotamos los **residuos**:

| División | Cociente | Residuo |
|---|---|---|
| 13 ÷ 2 | 6 | **1** |
| 6 ÷ 2 | 3 | **0** |
| 3 ÷ 2 | 1 | **1** |
| 1 ÷ 2 | 0 | **1** |

Leyendo los residuos **de abajo hacia arriba**: 13 = **1101**₂.

```python
def decimal_a_binario(n):
    """Convierte un entero n >= 0 a una cadena binaria."""
    if n == 0:
        return "0"
    cifras = ""
    while n > 0:
        cifras = str(n % 2) + cifras    # el residuo va a la IZQUIERDA de lo que llevamos
        n = n // 2
    return cifras

for n in [0, 1, 2, 5, 13, 100, 255, 2026]:
    print(n, "→", decimal_a_binario(n), "   (Python dice:", bin(n), ")")
```

```python
# Contemos en binario del 0 al 16. Fíjate en cómo "se lleva" igual que en base 10
for n in range(17):
    print(f"{n:3}  {decimal_a_binario(n):>6}")
```

## 3. Ver los bits

Pintemos los números del 0 al 63 como filas de 6 bits: negro = 1, blanco = 0.

```python
import matplotlib.pyplot as plt
import numpy as np

N = 64
bits = 6
tabla = np.zeros((N, bits))
for n in range(N):
    texto = decimal_a_binario(n).zfill(bits)     # zfill rellena con ceros a la izquierda
    for j in range(bits):
        tabla[n, j] = int(texto[j])

plt.figure(figsize=(4, 9))
plt.imshow(tabla, cmap="Greys", aspect="auto")
plt.xticks(range(bits), [str(2**(bits - 1 - j)) for j in range(bits)])
plt.xlabel("valor de la posición")
plt.ylabel("número n")
plt.title("Los números del 0 al 63 en binario")
plt.show()
```

- …cuentas cuántos números tienen el bit del **32** encendido? ¿Y el del 1?

### Un regalo de la Lección 2

Python puede comparar dos números bit a bit con el operador `&` ("y"): un bit del resultado es 1 solo si los dos números tienen 1 en esa posición. Pintemos de negro las parejas (a, b) con `a & b == 0`, es decir, que **no comparten ningún bit encendido**.

```python
N = 64
tabla = np.zeros((N, N))
for a in range(N):
    for b in range(N):
        if a & b == 0:
            tabla[b, a] = 1

plt.figure(figsize=(6, 6))
plt.imshow(tabla, cmap="Greys", origin="lower")
plt.xlabel("a"); plt.ylabel("b")
plt.title("Parejas (a, b) sin bits en común: a & b == 0")
plt.show()
```

¡Es el **triángulo de Sierpinski** otra vez! Esta vez no apareció por azar, sino por los bits. Cambia `N = 64` por `N = 128` o `N = 256` para ver más niveles.

## 4. El secreto de las tarjetas mágicas

Mira los números en binario. La **tarjeta 1** tiene todos los números con el bit del **1** encendido (los impares). La **tarjeta 2**, los que tienen encendido el bit del **2**. La tarjeta 3, el del **4**, y así: 1, 2, 4, 8, 16, 32.

El primer número de cada tarjeta es justamente ese valor. Si tu número está en las tarjetas 1, 3 y 4, en binario es 001101₂ = 1 + 4 + 8 = **13**.

```python
def tarjeta(t, maximo=63):
    """Números entre 1 y maximo que tienen encendido el bit de valor 2**(t-1)."""
    valor = 2**(t - 1)
    return [n for n in range(1, maximo + 1) if (n // valor) % 2 == 1]

def adivinar(tarjetas_elegidas):
    return sum(2**(t - 1) for t in tarjetas_elegidas)

# Comprobamos que el truco funciona para TODOS los números del 1 al 63
fallos = 0
for secreto in range(1, 64):
    elegidas = [t for t in range(1, 7) if secreto in tarjeta(t)]
    if adivinar(elegidas) != secreto:
        fallos = fallos + 1
print("Fallos del truco:", fallos)
```

Interactúa con el número: activa o desactiva cada bit con las casillas y mira qué número formas. (Es lo mismo que el truco: cada casilla es una tarjeta.)

```python
from ipywidgets import interact, Checkbox

def construir_numero(b32, b16, b8, b4, b2, b1):
    encendidos = [b32, b16, b8, b4, b2, b1]
    texto = "".join("1" if b else "0" for b in encendidos)
    valor = int(texto, 2)
    fig, ax = plt.subplots(figsize=(8, 1.6))
    for j, b in enumerate(encendidos):
        ax.add_patch(plt.Rectangle((j, 0), 0.9, 0.9, color="#f4a261" if b else "#e9ecef"))
        ax.text(j + 0.45, 0.45, "1" if b else "0", ha="center", va="center", fontsize=18)
        ax.text(j + 0.45, -0.25, str(2**(5 - j)), ha="center", va="center", fontsize=10, color="gray")
    ax.set_xlim(0, 6); ax.set_ylim(-0.5, 1); ax.axis("off")
    ax.set_title(f"{texto} en binario = {valor} en decimal")
    plt.show()

interact(construir_numero, b32=Checkbox(value=False, description="32"), b16=Checkbox(value=False, description="16"),
         b8=Checkbox(value=True, description="8"), b4=Checkbox(value=True, description="4"),
         b2=Checkbox(value=False, description="2"), b1=Checkbox(value=True, description="1"));
```

## 5. Otras bases

Nada obliga a usar 10 o 2. En **base b** cada posición vale **b veces** la anterior y las cifras van de 0 a b − 1. Generalicemos nuestra función:

```python
CIFRAS = "0123456789ABCDEF"     # para bases mayores que 10 usamos letras

def a_base(n, b):
    """Escribe el entero n >= 0 en base b (2 <= b <= 16)."""
    if n == 0:
        return "0"
    texto = ""
    while n > 0:
        texto = CIFRAS[n % b] + texto
        n = n // b
    return texto

for b in [2, 3, 5, 8, 10, 12, 16]:
    print(f"2026 en base {b:2}: {a_base(2026, b)}")
```

### Hexadecimal y colores

La base **16** (hexadecimal) usa las cifras 0–9 y A–F. Es muy útil porque **una cifra hexadecimal = 4 bits** exactos. En las páginas web, los colores se escriben como `#RRGGBB`: dos cifras hexadecimales para el rojo, dos para el verde y dos para el azul, cada una de 00 (nada) a FF = 255 (máximo).

```python
colores = ["#FF0000", "#00FF00", "#0000FF", "#FFFF00", "#FF8800", "#7B2CBF", "#808080", "#1F7A5C"]

fig, ax = plt.subplots(figsize=(10, 2))
for i, c in enumerate(colores):
    rojo, verde, azul = int(c[1:3], 16), int(c[3:5], 16), int(c[5:7], 16)
    ax.add_patch(plt.Rectangle((i, 0), 0.9, 1, color=c))
    ax.text(i + 0.45, -0.2, c, ha="center", fontsize=9)
    ax.text(i + 0.45, -0.45, f"{rojo},{verde},{azul}", ha="center", fontsize=8, color="gray")
ax.set_xlim(0, len(colores)); ax.set_ylim(-0.6, 1); ax.axis("off")
plt.show()
```

- …inventas tu propio color, por ejemplo `#20B2AA`? Agrégalo a la lista.
- …usas el mismo valor en las tres parejas (`#333333`, `#999999`, `#CCCCCC`)? ¿Qué colores salen?

## 6. ¿Cuántos bits necesita un número?

Cada bit extra **duplica** la cantidad de números que podemos escribir. Por eso los números de bits crecen muy despacio.

```python
numeros = range(1, 2001)
cantidad_bits = [len(decimal_a_binario(n)) for n in numeros]

plt.figure(figsize=(9, 4))
plt.step(numeros, cantidad_bits, where="post")
plt.xlabel("n"); plt.ylabel("cantidad de bits")
plt.title("¿Cuántos bits hacen falta para escribir n?")
plt.grid(True)
plt.show()

for n in [10, 1000, 10**6, 10**9, 10**100]:
    print(f"un número de {len(str(n))} cifras decimales necesita {len(decimal_a_binario(n))} bits")
```

La leyenda del ajedrez cuenta que su inventor pidió 1 grano de trigo por la primera casilla, 2 por la segunda, 4 por la tercera… duplicando hasta la casilla 64. En total: 1 + 2 + 4 + … + 2⁶³. En binario ese número son **64 unos seguidos**.

```python
total = sum(2**k for k in range(64))
print(total)
print(decimal_a_binario(total))
print("Cantidad de bits:", len(decimal_a_binario(total)))
```
