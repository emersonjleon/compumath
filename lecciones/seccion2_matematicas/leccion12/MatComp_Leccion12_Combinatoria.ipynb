## 📚 El concepto: tres preguntas antes de contar

La **combinatoria** cuenta posibilidades sin escribirlas todas. Antes de usar una fórmula, pregúntate:

| Pregunta | Si la respuesta es sí | Ejemplo |
|---|---|---|
| ¿Elijo en **etapas** independientes? | multiplica las opciones de cada etapa | camiseta × pantalón × zapatos |
| ¿Importa el **orden**? | permutaciones / variaciones | podio de una carrera |
| ¿El orden **no** importa? | combinaciones | elegir un equipo |

**Principio multiplicativo:** si una decisión tiene a opciones y otra tiene b opciones, juntas tienen **a · b** opciones.

## 1. El principio multiplicativo

```python
from itertools import product

camisetas = ["roja", "azul", "negra"]
pantalones = ["jean", "sudadera"]
zapatos = ["tenis", "botas"]

pintas = list(product(camisetas, pantalones, zapatos))   # todas las combinaciones de una de cada
for p in pintas:
    print(p)
print("Total:", len(pintas), "=", len(camisetas), "·", len(pantalones), "·", len(zapatos))
```

```python
# Contraseñas de 4 dígitos (del 0000 al 9999)
print("Contraseñas de 4 dígitos:", 10**4)
# Contraseñas de 8 caracteres con letras minúsculas, mayúsculas y dígitos
print("Contraseñas de 8 caracteres:", f"{62**8:,}")
```

## 2. Permutaciones: ordenar todo

¿De cuántas formas se pueden ordenar las letras A, B, C?

- Para el **primer** lugar hay 3 opciones.
- Para el **segundo**, quedan 2.
- Para el **tercero**, queda 1.

Total: 3 · 2 · 1 = 6. En general, n objetos se ordenan de **n! = n · (n − 1) · … · 2 · 1** formas (se lee "n factorial").

```python
from itertools import permutations
import math

for orden in permutations("ABC"):
    print("".join(orden))
print("Total:", len(list(permutations("ABC"))), "= 3! =", math.factorial(3))
```

```python
# ¡El factorial crece muy rápido!
for n in [1, 2, 3, 5, 10, 15, 20]:
    print(f"{n:2}! = {math.factorial(n):,}")

print()
print("52! =", math.factorial(52))
print("Tiene", len(str(math.factorial(52))), "cifras")
```

**52! ≈ 8 · 10⁶⁷**. Si cada persona de la Tierra hubiera barajado un mazo **cada segundo** desde el Big Bang, apenas habríamos probado una fracción minúscula de los órdenes posibles. ¡Casi seguro que tu mazo bien barajado tiene un orden que **nunca antes existió**!

```python
import matplotlib.pyplot as plt
import numpy as np

n = np.arange(1, 26)
plt.figure(figsize=(8, 4))
plt.semilogy(n, [math.factorial(k) for k in n], "o-", label="n!")
plt.semilogy(n, 2.0**n, "s-", label="2ⁿ")
plt.semilogy(n, n.astype(float)**3, "^-", label="n³")
plt.xlabel("n"); plt.legend(); plt.grid(True, alpha=0.3)
plt.title("El factorial le gana incluso a la exponencial (escala logarítmica)")
plt.show()
```

## 3. Variaciones: ordenar solo algunos

En una carrera de **8** corredores, ¿de cuántas formas puede quedar el **podio** (oro, plata, bronce)?

Para el oro hay 8 opciones, para la plata 7 y para el bronce 6: **8 · 7 · 6 = 336**. En general:

$$V(n, k) = n \cdot (n-1) \cdots (n-k+1) = \frac{n!}{(n-k)!}$$

```python
corredores = ["Ana", "Beto", "Caro", "Dani", "Eva", "Fede", "Gabi", "Hugo"]
podios = list(permutations(corredores, 3))       # ordenar 3 de los 8
print(podios[:5], "...")
print("Total:", len(podios), "   fórmula:", math.perm(8, 3))
```

## 4. Combinaciones: cuando el orden no importa

Ahora el profesor debe elegir **3** de los 8 para un comité. Ana-Beto-Caro es el **mismo** comité que Caro-Ana-Beto.

Cada comité aparece **3! = 6 veces** en la lista de podios (una por cada orden), así que:

$$C(n, k) = \binom{n}{k} = \frac{V(n,k)}{k!} = \frac{n!}{k!\,(n-k)!}$$

```python
from itertools import combinations

comites = list(combinations(corredores, 3))
print(comites[:5], "...")
print("Total:", len(comites), "   fórmula:", math.comb(8, 3), "  = 336 / 6 =", 336 // 6)
```

```python
# ¿Cuántos saludos de mano hay si cada una de n personas saluda a todas las demás?
for personas in [2, 3, 4, 10, 23, 30]:
    print(f"{personas:2} personas → {math.comb(personas, 2):3} saludos")
```

¿Te suenan 1, 3, 6, 45, 253? Son números **triangulares** (Lección 1), y 253 son las parejas de la paradoja del cumpleaños (Lección 10).

### ¿Qué tan difícil es ganar el Baloto?

En el Baloto (Colombia) se eligen **5 números distintos del 1 al 43** (el orden no importa) y además una **superbalota del 1 al 16**.

```python
formas = math.comb(43, 5) * 16
print(f"Formas de llenar un tiquete: {formas:,}")
print(f"Probabilidad de ganar con un tiquete: 1 en {formas:,} ≈ {1 / formas:.10f}")

# Si compras un tiquete cada semana, ¿cuántos años tardarías en promedio en ganar?
print(f"Años esperados comprando uno por semana: {formas / 52:,.0f}")
```

## 5. Caminos en una cuadrícula

Una hormiga va de la esquina inferior izquierda a la esquina superior derecha de una cuadrícula de **m × n**, dando solo pasos hacia la **derecha (D)** o hacia **arriba (A)**. ¿Cuántos caminos distintos hay?

Cada camino es una palabra con m letras D y n letras A, por ejemplo `DDADA`. Basta elegir **en qué posiciones van las A**: hay **C(m + n, n)** caminos.

Mueve los sliders para ver caminos al azar y el total.

```python
import random
from ipywidgets import interact, IntSlider

def ver_caminos(m, n, cuantos):
    fig, ax = plt.subplots(figsize=(6, 6 * n / m if m >= n else 6))
    for i in range(m + 1):
        ax.plot([i, i], [0, n], color="lightgray")
    for j in range(n + 1):
        ax.plot([0, m], [j, j], color="lightgray")
    rng = random.Random(1)
    for k in range(cuantos):
        pasos = ["D"] * m + ["A"] * n
        rng.shuffle(pasos)                    # un camino al azar
        x, y = [0], [0]
        for p in pasos:
            x.append(x[-1] + (p == "D"))
            y.append(y[-1] + (p == "A"))
        d = 0.25 * (k / max(cuantos - 1, 1) - 0.5)   # pequeño corrimiento para que no se tapen
        ax.plot(np.array(x) + d, np.array(y) + d, lw=2, alpha=0.8)
    ax.set_aspect("equal"); ax.axis("off")
    ax.set_title(f"Cuadrícula {m} × {n}: C({m + n}, {n}) = {math.comb(m + n, n):,} caminos")
    plt.show()

interact(ver_caminos,
         m=IntSlider(value=4, min=1, max=12, continuous_update=False),
         n=IntSlider(value=3, min=1, max=12, continuous_update=False),
         cuantos=IntSlider(value=3, min=1, max=20, continuous_update=False));
```

```python
# Comprobamos listando TODOS los caminos de una cuadrícula 4 × 3
caminos = set(permutations("DDDDAAA"))     # set quita los repetidos
print("Caminos distintos:", len(caminos), "   fórmula:", math.comb(7, 3))
print(["".join(c) for c in sorted(caminos)][:8], "...")
```

## 🎯 Cierre: lo que aprendiste hoy

- **Principio multiplicativo:** si se elige en etapas, se multiplican las opciones (pintas, contraseñas, placas).
- **Permutaciones:** n objetos se ordenan de **n!** formas; 52! tiene 68 cifras, así que cada mazo bien barajado es casi seguro único.
- **Variaciones:** ordenar k de n da n!/(n − k)! (el podio).
- **Combinaciones:** si el orden no importa, se divide entre k!: **C(n, k) = n!/(k!(n − k)!)** (comités, Baloto, caminos en una cuadrícula).
- En Python aprendiste: `itertools.product`, `permutations`, `combinations`, `combinations_with_replacement`, y `math.factorial`, `math.perm`, `math.comb`.
