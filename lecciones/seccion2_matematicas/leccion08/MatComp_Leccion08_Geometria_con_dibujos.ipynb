## 📚 El concepto: geometría de la tortuga

En los años 60, Seymour Papert creó el lenguaje **Logo** para que los niños aprendieran geometría dando órdenes a una tortuga robot que llevaba un lápiz. La tortuga solo sabe dos cosas:

| Orden | Qué hace |
|---|---|
| `avanzar(d)` | camina d pasos en la dirección hacia donde mira, dibujando una línea |
| `girar(a)` | gira a grados a la izquierda (sin moverse) |

Con eso basta para dibujar casi cualquier cosa. Para avanzar, la tortuga usa la trigonometría: si mira con un ángulo θ, se mueve d·cos(θ) en x y d·sin(θ) en y.

## 1. Nuestra tortuga

```python
import math
import matplotlib.pyplot as plt

class Tortuga:
    """Una tortuga que camina sobre el plano y recuerda su recorrido."""

    def __init__(self):
        self.x, self.y = 0.0, 0.0      # posición
        self.angulo = 0.0              # hacia dónde mira (0 = derecha)
        self.trazos = [[(0.0, 0.0)]]   # lista de líneas; cada línea es una lista de puntos
        self.giro_total = 0.0

    def avanzar(self, d):
        self.x += d * math.cos(math.radians(self.angulo))
        self.y += d * math.sin(math.radians(self.angulo))
        self.trazos[-1].append((self.x, self.y))

    def girar(self, a):
        self.angulo += a
        self.giro_total += a

    def saltar_a(self, x, y):
        """Se mueve sin dibujar y empieza una línea nueva."""
        self.x, self.y = x, y
        self.trazos.append([(x, y)])

    def dibujar(self, titulo="", color="#2a6f97", ax=None):
        if ax is None:
            fig, ax = plt.subplots(figsize=(6, 6))
        for trazo in self.trazos:
            xs = [p[0] for p in trazo]
            ys = [p[1] for p in trazo]
            ax.plot(xs, ys, color=color, lw=2)
        ax.scatter([self.x], [self.y], color="crimson", zorder=3)   # la tortuga
        ax.set_aspect("equal"); ax.axis("off")
        ax.set_title(titulo)
        return ax

t = Tortuga()
for _ in range(4):
    t.avanzar(10)
    t.girar(90)
t.dibujar("Un cuadrado: 4 veces (avanzar 10, girar 90°)")
plt.show()
print("Giro total:", t.giro_total, "grados")
```

Una **clase** (`class`) es un molde para crear objetos que guardan datos (la posición, el ángulo) y saben hacer cosas (avanzar, girar). `t = Tortuga()` crea una tortuga nueva, y `t.avanzar(10)` le da una orden.

Resolvamos la pregunta inicial:

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

t = Tortuga()
for _ in range(5):
    t.avanzar(10); t.girar(72)
t.dibujar(f"girar 72° → giro total {t.giro_total:.0f}°", ax=axes[0])

t = Tortuga()
for _ in range(5):
    t.avanzar(10); t.girar(144)
t.dibujar(f"girar 144° → giro total {t.giro_total:.0f}°", ax=axes[1], color="#e76f51")
plt.show()
```

Con 72° la tortuga dibuja un **pentágono** y gira en total 5 · 72° = **360°**: una vuelta completa. Con 144° dibuja una **estrella** y gira 5 · 144° = **720°**: dos vueltas.

**Teorema del giro total:** si la tortuga recorre una figura cerrada y termina mirando hacia donde empezó, el giro total es un **múltiplo de 360°**. Si la figura no se cruza a sí misma, es exactamente 360°.

## 2. Polígonos regulares

Para dibujar un polígono regular de **n** lados la tortuga debe girar en total 360°, repartidos en n giros iguales:

$$\text{ángulo exterior} = \frac{360°}{n} \qquad \text{ángulo interior} = 180° - \frac{360°}{n}$$

Mueve el slider para cambiar el número de lados.

```python
from ipywidgets import interact, IntSlider

def poligono(t, n, lado):
    for _ in range(n):
        t.avanzar(lado)
        t.girar(360 / n)

def ver_poligono(n):
    t = Tortuga()
    poligono(t, n, 30 / n)           # el perímetro siempre mide 30
    exterior = 360 / n
    interior = 180 - exterior
    t.dibujar(f"n = {n}: ángulo exterior {exterior:.1f}°, interior {interior:.1f}°")
    plt.show()
    print(f"Suma de los ángulos interiores: {n} · {interior:.2f}° = {n * interior:.0f}°")

interact(ver_poligono, n=IntSlider(value=6, min=3, max=40, continuous_update=False));
```

## 3. Estrellas {n/k}

Si en lugar de girar 360°/n la tortuga gira **k · 360°/n**, salta k vértices cada vez y dibuja la **estrella {n/k}**. A veces sale de un solo trazo y a veces se queda en una parte de la figura. Explora con los sliders.

```python
def estrella(n, k, lado=10):
    """Dibuja la estrella {n/k} y devuelve la tortuga."""
    t = Tortuga()
    pasos = 0
    while True:
        t.avanzar(lado)
        t.girar(360 * k / n)
        pasos += 1
        # paramos cuando la tortuga vuelve al inicio mirando igual
        if abs(t.x) < 1e-6 and abs(t.y) < 1e-6 and round(t.giro_total) % 360 == 0:
            break
    return t, pasos

def ver_estrella(n, k):
    t, pasos = estrella(n, k)
    t.dibujar(f"{{{n}/{k}}}: {pasos} lados, giro total {t.giro_total:.0f}° = {round(t.giro_total / 360)} vueltas")
    plt.show()
    print(f"La tortuga visitó {pasos} de los {n} vértices")
    print(f"MCD({n}, {k}) = {math.gcd(n, k)}")

interact(ver_estrella,
         n=IntSlider(value=7, min=3, max=30, continuous_update=False),
         k=IntSlider(value=3, min=1, max=14, continuous_update=False));
```

**¿Qué pasa si…?**
- …pruebas {8/3} y {8/2}? ¿Por qué una sale completa y la otra no?
- …comparas el número de vértices visitados con n / MCD(n, k)? ¿Coinciden siempre?
- …{n/k} y {n/(n−k)}? ¿Dibujan lo mismo?

La estrella sale de **un solo trazo** exactamente cuando **MCD(n, k) = 1**. Si el MCD es d, la tortuga solo visita n/d vértices; para completar la figura hay que dibujar d estrellas giradas.

```python
# Estrellas completas: si MCD(n, k) = d, dibujamos d copias giradas
def estrella_completa(n, k, ax):
    d = math.gcd(n, k)
    colores = ["#2a6f97", "#e76f51", "#2a9d8f", "#f4a261", "#7b2cbf", "#e63946"]
    for i in range(d):
        t, _ = estrella(n, k)
        # centro de la estrella (la tortuga empieza en un vértice mirando a la derecha)
        cx = 10 / 2
        cy = (10 / 2) / math.tan(math.pi * k / n)
        # rotamos la copia i un ángulo 360·i/n alrededor de ese centro
        a = 2 * math.pi * i / n
        for trazo in t.trazos:
            xs = [cx + (p[0] - cx) * math.cos(a) - (p[1] - cy) * math.sin(a) for p in trazo]
            ys = [cy + (p[0] - cx) * math.sin(a) + (p[1] - cy) * math.cos(a) for p in trazo]
            ax.plot(xs, ys, color=colores[i % len(colores)], lw=2)
    ax.set_aspect("equal"); ax.axis("off")
    ax.set_title(f"{{{n}/{k}}}: MCD = {d}, {d} trazo(s)")

fig, axes = plt.subplots(1, 3, figsize=(13, 4.5))
estrella_completa(6, 2, axes[0])
estrella_completa(10, 4, axes[1])
estrella_completa(12, 5, axes[2])
plt.show()
```

## 4. Polígonos y π: el método de Arquímedes

Hace 2200 años, Arquímedes calculó π usando polígonos regulares **inscritos** en un círculo de diámetro 1: cuantos más lados, más se parece el perímetro del polígono a la longitud de la circunferencia, que es π.

Un polígono de n lados inscrito en un círculo de radio 1/2 tiene lados de longitud sin(180°/n).

```python
print(" lados   perímetro       error")
for n in [6, 12, 24, 48, 96, 1000, 10**6]:
    perimetro = n * math.sin(math.pi / n)
    print(f"{n:7}   {perimetro:.10f}   {math.pi - perimetro:.2e}")
print("      π  ", math.pi)
```

```python
fig, axes = plt.subplots(1, 4, figsize=(14, 4))
for ax, n in zip(axes, [6, 12, 24, 96]):
    circulo = plt.Circle((0, 0), 0.5, fill=False, color="lightgray", lw=3)
    ax.add_patch(circulo)
    xs = [0.5 * math.cos(2 * math.pi * i / n) for i in range(n + 1)]
    ys = [0.5 * math.sin(2 * math.pi * i / n) for i in range(n + 1)]
    ax.plot(xs, ys, color="#2a6f97")
    ax.set_aspect("equal"); ax.axis("off")
    ax.set_title(f"n = {n}: {n * math.sin(math.pi / n):.5f}")
plt.show()
```

Arquímedes llegó hasta **96 lados** (¡a mano!) y concluyó que 3 + 10/71 < π < 3 + 1/7, es decir, π ≈ 3,14.

## 5. Espirales y rosetas

Si la tortuga **avanza un poco más** en cada paso, en vez de un polígono dibuja una **espiral**. Mueve el ángulo: valores un poco distintos de 90° o 120° dan efectos sorprendentes.

```python
from ipywidgets import FloatSlider

def espiral(angulo, pasos):
    t = Tortuga()
    for i in range(pasos):
        t.avanzar(i)          # cada lado es un poco más largo
        t.girar(angulo)
    fig, ax = plt.subplots(figsize=(7, 7))
    t.dibujar(f"Espiral: girar {angulo}° en cada paso", ax=ax, color="#7b2cbf")
    for linea in ax.lines:
        linea.set_linewidth(0.8)
    plt.show()

interact(espiral,
         angulo=FloatSlider(value=91, min=30, max=179, step=1, continuous_update=False),
         pasos=IntSlider(value=150, min=10, max=400, continuous_update=False));
```

**¿Qué pasa si…?**
- …pruebas 90°, 89° y 91°? ¿Por qué la espiral "rota" hacia un lado o hacia el otro?
- …pruebas 120°, 121° y 144°?

Una **roseta** se forma repitiendo un polígono mientras la tortuga gira un poco entre copia y copia:

```python
def roseta(lados, copias):
    t = Tortuga()
    for _ in range(copias):
        poligono(t, lados, 10)
        t.girar(360 / copias)
    fig, ax = plt.subplots(figsize=(6, 6))
    t.dibujar(f"Roseta: {copias} {lados}-ágonos", ax=ax, color="#2a9d8f")
    for linea in ax.lines:
        linea.set_linewidth(1)
    plt.show()

interact(roseta,
         lados=IntSlider(value=5, min=3, max=10, continuous_update=False),
         copias=IntSlider(value=12, min=1, max=36, continuous_update=False));
```
