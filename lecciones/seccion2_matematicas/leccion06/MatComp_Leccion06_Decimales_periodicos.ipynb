## 📚 El concepto: decimales finitos y periódicos

Al dividir el numerador entre el denominador puede pasar una de dos cosas:

| Tipo | Ejemplo | Qué pasa |
|---|---|---|
| **Finito** | 3/8 = 0,375 | La división llega a residuo 0 y se acaba. |
| **Periódico** | 4/11 = 0,363636… = 0,$\overline{36}$ | Un bloque de cifras (el **periodo**) se repite para siempre. |

Hoy descubriremos que **no hay otra opción**: toda fracción es finita o periódica. Y al revés: todo decimal periódico es una fracción.

## 2. La división larga

Así dividimos 1 entre 7 a mano:

| Paso | Residuo × 10 | Cifra (÷ 7) | Nuevo residuo |
|---|---|---|---|
| 1 | 10 | **1** | 3 |
| 2 | 30 | **4** | 2 |
| 3 | 20 | **2** | 6 |
| 4 | 60 | **8** | 4 |
| 5 | 40 | **5** | 5 |
| 6 | 50 | **7** | 1 |
| 7 | 10 | **1** | 3 ← ¡otra vez el residuo del paso 1! |

En cada paso: **cifra = (residuo · 10) // d** y **nuevo residuo = (residuo · 10) % d**.

```python
def division_larga(a, d, cifras=30):
    """Devuelve las primeras cifras decimales de a/d (con 0 <= a < d) como texto."""
    texto = ""
    r = a
    for _ in range(cifras):
        r = r * 10
        texto = texto + str(r // d)   # la cifra que escribimos
        r = r % d                     # lo que sobra para el siguiente paso
    return "0," + texto

print("1/7  =", division_larga(1, 7))
print("1/17 =", division_larga(1, 17, 40))
print("3/8  =", division_larga(3, 8))
print("1/97 =", division_larga(1, 97, 100))
```

```python
# El número cíclico: los múltiplos de 1/7
for a in range(1, 7):
    print(f"{a}/7 =", division_larga(a, 7, 18))
```

## 3. ¿Por qué se repiten? Mirando los residuos

Al dividir entre d, el residuo siempre está entre 0 y d − 1. Solo hay **d** residuos posibles.

- Si alguna vez sale **0**, la división termina: decimal **finito**.
- Si nunca sale 0, después de como mucho d − 1 pasos **un residuo se repite**… y desde ahí todo se repite igual. Es el **principio del palomar**: si metes más palomas que nidos, algún nido tiene dos palomas.

```python
def residuos(a, d, pasos=20):
    """Lista de residuos de la división larga de a entre d."""
    lista = []
    r = a
    for _ in range(pasos):
        lista.append(r)
        r = (r * 10) % d
    return lista

print("Residuos de 1/7: ", residuos(1, 7))
print("Residuos de 3/8: ", residuos(3, 8))
print("Residuos de 1/12:", residuos(1, 12))
```

```python
def periodo(a, d):
    """Longitud del periodo de a/d (0 si el decimal es finito)."""
    vistos = {}           # diccionario: residuo -> paso en que apareció
    r = a % d
    paso = 0
    while r != 0 and r not in vistos:
        vistos[r] = paso
        r = (r * 10) % d
        paso = paso + 1
    if r == 0:
        return 0
    return paso - vistos[r]

for d in [3, 6, 7, 8, 11, 12, 13, 17, 97]:
    print(f"1/{d}: periodo {periodo(1, d)}")
```

Usamos un **diccionario** (`{}`): guarda parejas *clave → valor*. Aquí la clave es el residuo y el valor es el paso en el que apareció; así sabemos dónde empieza el periodo.

¿Acertaste tu predicción? 1/17 tiene periodo **16**, más largo que el 6 de 1/7.

## 4. Periodos de 1/n

Calculemos el periodo de 1/n para n del 2 al 300 y grafiquemos.

```python
import matplotlib.pyplot as plt

def es_primo(n):
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d = d + 1
    return True

ns = list(range(2, 301))
periodos = [periodo(1, n) for n in ns]
colores = ["crimson" if es_primo(n) else "gray" for n in ns]

plt.figure(figsize=(10, 5))
plt.scatter(ns, periodos, s=10, c=colores)
plt.plot([2, 300], [1, 299], "--", color="lightgray", label="periodo = n − 1")
plt.xlabel("n"); plt.ylabel("periodo de 1/n")
plt.title("Periodo del decimal de 1/n (rojo = n primo)")
plt.legend()
plt.show()
```

Observa:
- Muchos puntos rojos están **justo sobre la recta** periodo = n − 1: son primos con periodo **máximo**, como 7, 17, 19, 23, 29… (se llaman *primos de periodo completo*).
- Hay puntos en **cero**: son los decimales finitos. ¿Qué números son?

```python
finitos = [n for n in ns if periodo(1, n) == 0]
print("1/n es finito para n =", finitos)
```

¡Solo aparecen números de la forma **2ᵃ · 5ᵇ**! Como 10 = 2 · 5, una fracción (simplificada) es finita exactamente cuando su denominador no tiene más factores primos que 2 y 5.

## 5. Dibujando decimales en un círculo

Coloca las cifras 0, 1, …, 9 alrededor de un círculo, como un reloj. Ahora recorre las cifras del decimal de a/d y une con una línea cada cifra con la siguiente. Como el decimal es periódico, el dibujo se **cierra**.

Usa los **sliders** para cambiar el numerador y el denominador.

```python
import math
from ipywidgets import interact, IntSlider

def posicion(cifra):
    angulo = math.pi / 2 - 2 * math.pi * cifra / 10   # 0 arriba, en sentido del reloj
    return math.cos(angulo), math.sin(angulo)

def dibujar_decimal(numerador, denominador):
    a = numerador % denominador
    p = periodo(a, denominador)
    n_cifras = 60 if p == 0 else 2 * p + 20
    texto = division_larga(a, denominador, n_cifras)[2:]
    xs = [posicion(int(c))[0] for c in texto]
    ys = [posicion(int(c))[1] for c in texto]

    fig, ax = plt.subplots(figsize=(6, 6))
    circulo = plt.Circle((0, 0), 1, fill=False, color="lightgray")
    ax.add_patch(circulo)
    ax.plot(xs, ys, color="#2a6f97", lw=1.5, alpha=0.8)
    for c in range(10):
        x, y = posicion(c)
        ax.text(1.12 * x, 1.12 * y, str(c), ha="center", va="center", fontsize=14)
    ax.set_xlim(-1.3, 1.3); ax.set_ylim(-1.3, 1.3)
    ax.set_aspect("equal"); ax.axis("off")
    tipo = "finito" if p == 0 else f"periodo {p}"
    ax.set_title(f"{a}/{denominador} = 0,{texto[:16]}…  ({tipo})")
    plt.show()

interact(dibujar_decimal,
         numerador=IntSlider(value=1, min=1, max=60, continuous_update=False),
         denominador=IntSlider(value=7, min=2, max=200, continuous_update=False));
```

- …pruebas 1/13 y 2/13? ¿Son iguales los dibujos?
- …buscas un denominador cuyo dibujo use **las diez cifras**? (Prueba 1/17, 1/97…)

## 6. Del decimal periódico a la fracción

¿Qué fracción es x = 0,363636…? El truco: multiplicamos por 100 (porque el periodo tiene 2 cifras) y **restamos**:

$$\begin{aligned} 100x &= 36{,}363636\ldots \\ x &= \phantom{3}0{,}363636\ldots \\ \hline 99x &= 36 \end{aligned} \quad\Rightarrow\quad x = \frac{36}{99} = \frac{4}{11}$$

Python tiene el módulo `fractions` para trabajar con fracciones **exactas**:

```python
from fractions import Fraction

x = Fraction(36, 99)
print(x)                       # se simplifica sola (¡usa el MCD de la Lección 4!)
print(division_larga(4, 11, 12))

def periodico_a_fraccion(bloque):
    """Convierte 0,(bloque)(bloque)... en fracción. Ejemplo: '36' -> 4/11."""
    return Fraction(int(bloque), int("9" * len(bloque)))

for bloque in ["3", "36", "142857", "9"]:
    print(f"0,{bloque}{bloque}{bloque}... =", periodico_a_fraccion(bloque))
```

¡La última línea dice que **0,999… = 1**!

Mucha gente piensa que 0,999… es "un poquito menos" que 1. Pero:
- 1/3 = 0,333…, y 3 · (1/3) = 1, mientras que 3 · 0,333… = 0,999….
- Con el truco: 10x − x = 9,999… − 0,999… = 9, así que 9x = 9 y x = 1.
- Si 0,999… fuera menor que 1, habría un número entre los dos. ¿Cuál sería?

No son "casi iguales": son **dos formas de escribir el mismo número**.

```python
# Las sumas parciales 0,9 + 0,09 + 0,009 + ... se acercan a 1 tanto como queramos
suma = Fraction(0)
for k in range(1, 11):
    suma = suma + Fraction(9, 10**k)
    print(f"{k:2} nueves: {float(suma):.12f}   le falta {1 - suma}")
```

## 🎯 Cierre: lo que aprendiste hoy

- La **división larga** se programa con dos operaciones: `cifra = (r·10) // d` y `r = (r·10) % d`.
- Toda fracción da un decimal **finito** (si aparece residuo 0) o **periódico** (si un residuo se repite): solo hay d residuos posibles (**principio del palomar**).
- 1/n es **finito** exactamente cuando n = 2ᵃ · 5ᵇ; el periodo de 1/p puede llegar a p − 1 (**primos de periodo completo** como 7 y 17).
- Todo decimal periódico es una **fracción**: 0,$\overline{36}$ = 36/99 = 4/11, y **0,999… = 1**.
- La computadora guarda los decimales en binario con bits limitados, por eso **0.1 + 0.2 ≠ 0.3**.
- En Python aprendiste: diccionarios (`{}`), el módulo `fractions` y dibujos interactivos con sliders.
