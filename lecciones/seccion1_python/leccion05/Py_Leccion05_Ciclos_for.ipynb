## 1. Repetir con `for`

Si quisieras mostrar un mensaje 5 veces, podrías copiar la línea 5 veces. Pero ¿y si fueran 1000?

Un **ciclo `for`** repite un bloque de instrucciones. Como con el `if`, la línea termina en **dos puntos** y el bloque va **indentado**.

````python
for i in range(5):
    print("¡Voy a aprender Python!")
````

En cada vuelta, la variable `i` toma un valor distinto. Veámoslo:

````python
for i in range(5):
    print("Vuelta número", i)
````

`range(5)` produce los números **0, 1, 2, 3, 4**: empieza en 0 y llega **hasta antes de** 5, como las rebanadas de la Lección 3. Son 5 vueltas.

## 2. `range` con inicio, fin y paso

| Escribes | Números que produce |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(0, 20, 5)` | 0, 5, 10, 15 |
| `range(10, 0, -1)` | 10, 9, 8, …, 1 |

💡 `print(..., end=" ")` muestra el resultado con un espacio al final en vez de saltar de línea. Así vemos los números uno al lado del otro.

````python
for n in range(1, 11):
    print(n, end=" ")
````

````python
for n in range(0, 50, 7):      # de 7 en 7
    print(n, end=" ")
````

````python
for n in range(10, 0, -1):     # hacia atrás
    print(n, end=" ")
print("¡Despegue! 🚀")
````

## 3. Acumuladores: sumar dentro de un ciclo

Para sumar muchos números se usa una variable que **acumula**:
1. **Antes** del ciclo, se empieza en 0.
2. **Dentro** del ciclo, se le suma cada número.
3. **Después** del ciclo, se muestra el resultado.

````python
suma = 0                     # 1. empezar en cero
for n in range(1, 101):      # n = 1, 2, ..., 100
    suma = suma + n          # 2. sumar cada número (también: suma += n)
print("1 + 2 + ... + 100 =", suma)    # 3. mostrar al final
````

### 🎯 ¿Cómo lo hizo Gauss?

Gauss emparejó el primero con el último: 1 + 100 = 101, 2 + 99 = 101, 3 + 98 = 101… Hay **50 parejas** que suman 101, así que el total es 50 × 101 = **5050**.

En general, 1 + 2 + … + n = n × (n + 1) / 2. Comprobemos con un millón:

````python
n = 1000000

suma = 0
for k in range(1, n + 1):
    suma += k

print("Con el ciclo:   ", suma)
print("Con la fórmula: ", n * (n + 1) // 2)
````

El computador hizo un millón de sumas en menos de un segundo. ¡Pero Gauss encontró un atajo que sirve para cualquier n!

### Multiplicar en un ciclo: el factorial

El **factorial** de n, que se escribe n!, es 1 × 2 × 3 × … × n. Ojo: para multiplicar, el acumulador empieza en **1**, no en 0. (¿Por qué?)

````python
n = 10
producto = 1
for k in range(1, n + 1):
    producto *= k
print(f"{n}! = {producto}")
````

### Contadores

Un **contador** es un acumulador que suma 1 cada vez que pasa algo. Por ejemplo, ¿cuántos múltiplos de 7 hay entre 1 y 1000?

````python
contador = 0
for n in range(1, 1001):
    if n % 7 == 0:           # ¡un if dentro de un for!
        contador += 1
print("Múltiplos de 7 entre 1 y 1000:", contador)
````

## 4. Recorrer un texto

`for` también recorre un texto, **letra por letra**.

````python
for letra in "Python":
    print(letra)
````

````python
# Contar las vocales de una frase
frase = "Me gusta programar en Colab"
vocales = 0
for letra in frase.lower():
    if letra in "aeiouáéíóú":
        vocales += 1
print(f"La frase tiene {vocales} vocales")
````

````python
# Construir un texto nuevo: el acumulador empieza como texto vacío ""
palabra = "murciélago"
con_guiones = ""
for letra in palabra:
    con_guiones += letra + "-"
print(con_guiones)
````

## 5. `for` con `if`: el juego FizzBuzz

Un juego clásico para practicar: se cuenta desde 1, pero
- los múltiplos de 3 se reemplazan por **Fizz**,
- los múltiplos de 5 por **Buzz**,
- y los múltiplos de 3 y 5 a la vez por **FizzBuzz**.

````python
for n in range(1, 21):
    if n % 3 == 0 and n % 5 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
````

¿Por qué la condición de **FizzBuzz** va primero? Piensa en lo que aprendiste del orden de los `elif` en la Lección 4.

## 6. Ciclos dentro de ciclos

Un `for` puede ir dentro de otro. Por **cada** vuelta del ciclo de afuera, el de adentro da **todas** sus vueltas.

````python
for fila in range(1, 6):
    for col in range(1, 6):
        print(f"{fila * col:4}", end="")     # :4 reserva 4 espacios para alinear
    print()                                  # salto de línea al terminar cada fila
````

````python
# Un triángulo de estrellas: "*" * n repite el texto n veces (Lección 3)
for n in range(1, 6):
    print("⭐" * n)
````

## 7. Dibujar con un ciclo

En cada vuelta podemos dibujar un punto. Aquí comparamos la suma 1 + 2 + … + n (los **números triangulares**) con n². No te preocupes por los detalles de `matplotlib`; lo verás en la Lección 10.

````python
import matplotlib.pyplot as plt

suma = 0
for n in range(1, 21):
    suma += n
    plt.scatter(n, suma, color="tab:blue")
    plt.scatter(n, n ** 2, color="tab:orange", marker="s")

plt.title("Azul: 1 + 2 + ... + n      Naranja: n²")
plt.xlabel("n")
plt.grid(alpha=0.3)
plt.show()
````

¿Ves que los puntos azules quedan más o menos a la **mitad** de los naranjas? Eso coincide con la fórmula de Gauss: n(n + 1)/2 ≈ n²/2.

## 8. Errores comunes

````python
# Error: range solo acepta números enteros
for i in range(5.5):
    print(i)
````

````python
# Error de lógica: el acumulador se reinicia en cada vuelta
for n in range(1, 101):
    suma = 0            # ¡está DENTRO del ciclo!
    suma += n
print(suma)             # debería dar 5050...
````

````python
# Error de lógica: olvidar que range llega hasta ANTES del fin
suma = 0
for n in range(1, 100):     # ¡no incluye el 100!
    suma += n
print(suma)                 # da 4950 en vez de 5050
````

Estos dos últimos no muestran ningún mensaje de error, pero el resultado está mal. Por eso siempre conviene **comprobar** con un caso pequeño que conozcas.
