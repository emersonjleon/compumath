## 1. Tu primera gráfica

`matplotlib` es la librería de gráficas más usada en Python. Se importa con el apodo `plt`.

`plt.plot(x, y)` une con líneas los puntos `(x[0], y[0])`, `(x[1], y[1])`, … Las dos listas deben tener **la misma cantidad** de elementos.

````python
import matplotlib.pyplot as plt

dias = [1, 2, 3, 4, 5, 6, 7]
temperatura = [18, 20, 23, 22, 19, 21, 24]

plt.plot(dias, temperatura)
plt.show()
````

Funciona, pero una gráfica sin título ni ejes no le dice nada a nadie. Agreguemos información:

````python
plt.plot(dias, temperatura, color="tomato", marker="o", linestyle="--")
plt.title("Temperatura máxima de la semana")
plt.xlabel("día")
plt.ylabel("temperatura (°C)")
plt.grid(alpha=0.3)        # cuadrícula suave (alpha = transparencia)
plt.show()
````

### Opciones de estilo más comunes

| Opción | Ejemplos |
|---|---|
| `color=` | `"red"`, `"tab:blue"`, `"#5b8def"` |
| `marker=` | `"o"` círculo, `"s"` cuadrado, `"^"` triángulo, `"*"` estrella |
| `linestyle=` | `"-"` continua, `"--"` guiones, `":"` puntos, `""` sin línea |
| `linewidth=` | `1`, `2`, `3`… |
| `label=` | el nombre que aparece en la leyenda |

## 2. Graficar funciones

Para dibujar una función como f(x) = x², calculamos muchos valores de x muy juntitos y sus imágenes. Una comprensión de lista lo hace fácil.

````python
xs = [i / 10 for i in range(-30, 31)]       # -3.0, -2.9, ..., 3.0
ys = [x ** 2 for x in xs]

plt.plot(xs, ys)
plt.title("f(x) = x²")
plt.grid(alpha=0.3)
plt.show()
````

### Varias funciones en la misma figura

Cada `plt.plot` agrega una línea. Con `label=` y `plt.legend()` aparece una leyenda.

````python
def lineal(x):
    return 2 * x + 1

def cuadratica(x):
    return x ** 2

def cubica(x):
    return x ** 3 / 3

xs = [i / 10 for i in range(-30, 31)]

plt.plot(xs, [lineal(x) for x in xs], label="2x + 1")
plt.plot(xs, [cuadratica(x) for x in xs], label="x²")
plt.plot(xs, [cubica(x) for x in xs], label="x³ / 3")
plt.axhline(0, color="black", linewidth=0.8)      # eje x
plt.axvline(0, color="black", linewidth=0.8)      # eje y
plt.ylim(-6, 10)                                  # límites del eje y
plt.legend()
plt.grid(alpha=0.3)
plt.title("Tres funciones")
plt.show()
````

### Funciones trigonométricas

El módulo `math` trae `sin`, `cos`, `sqrt`, `pi` y muchas más.

````python
import math

xs = [i / 20 for i in range(0, 127)]       # de 0 a 6.3 ≈ 2π
plt.plot(xs, [math.sin(x) for x in xs], label="seno")
plt.plot(xs, [math.cos(x) for x in xs], label="coseno")
plt.xticks([0, math.pi / 2, math.pi, 3 * math.pi / 2, 2 * math.pi],
           ["0", "π/2", "π", "3π/2", "2π"])
plt.legend()
plt.grid(alpha=0.3)
plt.title("Seno y coseno")
plt.show()
````

## 3. Otros tipos de gráficas

### Dispersión: `plt.scatter`

Pone un punto por cada dato **sin unirlos**. Sirve para ver si dos cosas están relacionadas.

````python
horas_estudio = [1, 2, 2, 3, 4, 5, 6, 6, 7, 8, 9, 10]
nota = [2.1, 2.4, 2.9, 2.7, 3.2, 3.0, 3.6, 3.9, 3.8, 4.1, 4.4, 4.6]

plt.scatter(horas_estudio, nota, color="purple", s=60)     # s = tamaño de los puntos
plt.xlabel("horas de estudio")
plt.ylabel("nota")
plt.title("¿Estudiar más sube la nota?")
plt.grid(alpha=0.3)
plt.show()
````

### Barras: `plt.bar`

Para comparar cantidades de categorías. ¡Combina muy bien con los diccionarios de la lección anterior!

````python
encuesta = {"fútbol": 12, "baloncesto": 5, "natación": 4, "ciclismo": 7, "ajedrez": 3}

plt.bar(encuesta.keys(), encuesta.values(), color="#5b8def")
plt.title("Deporte favorito del curso")
plt.ylabel("número de estudiantes")
plt.show()
````

### Histograma: `plt.hist`

Agrupa muchos números en intervalos y cuenta cuántos caen en cada uno. Lancemos dos dados 5000 veces (como en la Lección 7):

````python
import random

sumas = [random.randint(1, 6) + random.randint(1, 6) for _ in range(5000)]

plt.hist(sumas, bins=range(2, 14), color="#a6da95", edgecolor="white", align="left")
plt.xticks(range(2, 13))
plt.xlabel("suma de los dos dados")
plt.ylabel("veces")
plt.title("5000 lanzamientos de dos dados")
plt.show()
````

💡 En `for _ in range(5000)`, el guion bajo `_` es un nombre de variable que se usa cuando **no nos importa** su valor.

## 4. Varias gráficas en una figura: `subplots`

`fig, ejes = plt.subplots(filas, columnas)` crea una cuadrícula de gráficas. Cada una se maneja con `ejes[i]`, y los nombres cambian un poco: `plt.title` se vuelve `eje.set_title`, `plt.xlabel` se vuelve `eje.set_xlabel`, etc.

````python
fig, ejes = plt.subplots(1, 3, figsize=(14, 4))       # 1 fila, 3 columnas; figsize = tamaño

ejes[0].plot(dias, temperatura, marker="o")
ejes[0].set_title("Líneas")

ejes[1].scatter(horas_estudio, nota)
ejes[1].set_title("Dispersión")

ejes[2].bar(encuesta.keys(), encuesta.values())
ejes[2].set_title("Barras")
ejes[2].tick_params(axis="x", rotation=45)

plt.tight_layout()          # acomoda los espacios para que nada se monte
plt.show()
````

## 5. Curvas paramétricas

En vez de y = f(x), una curva **paramétrica** da x **y** y en función de un tercer número `t`. Así se pueden dibujar figuras que no son funciones, como un círculo:

$$x = \cos(t) \qquad y = \sin(t) \qquad 0 \le t \le 2\pi$$

````python
ts = [2 * math.pi * i / 200 for i in range(201)]

plt.plot([math.cos(t) for t in ts], [math.sin(t) for t in ts])
plt.axis("equal")            # misma escala en x y en y, para que el círculo no salga ovalado
plt.title("Un círculo paramétrico")
plt.grid(alpha=0.3)
plt.show()
````

### 🎯 Volvamos al corazón

````python
ts = [2 * math.pi * i / 400 for i in range(401)]
xs = [16 * math.sin(t) ** 3 for t in ts]
ys = [13 * math.cos(t) - 5 * math.cos(2 * t) - 2 * math.cos(3 * t) - math.cos(4 * t) for t in ts]

plt.figure(figsize=(5, 5))
plt.fill(xs, ys, color="crimson")         # fill rellena el interior de la curva
plt.axis("equal")
plt.axis("off")                           # sin ejes
plt.title("❤ Hecho con ecuaciones")
plt.show()
````

## 6. Gráficas interactivas

Con un *slider* podemos ver cómo cambia una gráfica al mover un número. Mueve `a` (la altura) y `b` (qué tan rápido oscila) en y = a·sin(b·x).

````python
from ipywidgets import interact, FloatSlider

def onda(a=1.0, b=1.0):
    xs = [i / 20 for i in range(0, 127)]
    plt.plot(xs, [a * math.sin(b * x) for x in xs])
    plt.ylim(-3.2, 3.2)
    plt.title(f"y = {a:.1f} · sin({b:.1f} · x)")
    plt.grid(alpha=0.3)
    plt.show()

interact(onda,
         a=FloatSlider(value=1.0, min=0.1, max=3.0, step=0.1, continuous_update=False),
         b=FloatSlider(value=1.0, min=0.5, max=5.0, step=0.5, continuous_update=False));
````

````python
# Error de "olvido": sin plt.legend() las etiquetas no aparecen
plt.plot([1, 2, 3], [1, 4, 9], label="cuadrados")
plt.title("¿Dónde está la leyenda?")
plt.show()
````
