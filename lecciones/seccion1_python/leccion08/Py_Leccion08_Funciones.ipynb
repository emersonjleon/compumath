## 1. Tu primera función

Ya has usado muchas funciones: `print()`, `len()`, `sum()`, `input()`… Ahora vas a crear las tuyas.

```python
def nombre_de_la_funcion():
    instrucciones
```

- `def` (de *define*) indica que vamos a crear una función.
- Después van el nombre, paréntesis y **dos puntos**.
- El cuerpo va **indentado**, como en `if` y `for`.

Definir una función **no la ejecuta**: solo le enseña a Python una nueva instrucción. Para ejecutarla hay que **llamarla** escribiendo su nombre con paréntesis.

````python
def saludar():
    print("¡Hola!")
    print("Bienvenido a la lección de funciones.")

# Hasta aquí no pasa nada. Ahora la llamamos dos veces:
saludar()
saludar()
````

## 2. Parámetros: darle información a la función

Entre los paréntesis se ponen **parámetros**: variables que reciben los valores que le pasas al llamarla.

````python
def saludar_a(nombre):
    print(f"¡Hola, {nombre}!")

saludar_a("Ana")
saludar_a("Luis")
````

````python
def tabla(numero, hasta):
    for k in range(1, hasta + 1):
        print(f"{numero} x {k} = {numero * k}")

tabla(7, 5)
````

## 3. `return`: devolver un resultado

Las funciones más útiles **calculan algo y lo devuelven** con `return`, para que lo puedas guardar en una variable o usarlo en otra cuenta, igual que haces con `len()` o `sum()`.

````python
def area_rectangulo(base, altura):
    return base * altura

a = area_rectangulo(8, 5)
print("Área:", a)
print("Dos rectángulos juntos:", area_rectangulo(8, 5) + area_rectangulo(3, 2))
````

### `print` no es lo mismo que `return`

Esta es la confusión más común. Compara estas dos funciones:

````python
def doble_con_print(x):
    print(x * 2)          # solo MUESTRA el resultado en pantalla

def doble_con_return(x):
    return x * 2          # DEVUELVE el resultado

a = doble_con_print(5)
b = doble_con_return(5)
print("a vale:", a)
print("b vale:", b)
````

La primera mostró el 10, pero **no lo devolvió**: por eso `a` vale `None` ("nada"). La segunda no mostró nada, pero `b` sí guarda el 10.

💡 Regla práctica: si quieres **usar** el resultado después, usa `return`.

Además, `return` **termina** la función en ese momento, aunque haya más líneas después.

````python
def es_primo(n):
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False      # encontró un divisor: termina de una vez
    return True               # si llegó hasta aquí, no tiene divisores

print(es_primo(7), es_primo(12), es_primo(97))

# ¡Y ahora se puede usar en una comprensión de lista!
print([n for n in range(1, 50) if es_primo(n)])
````

## 4. Valores por defecto

Un parámetro puede tener un valor **por defecto**, que se usa si no le pasas nada.

````python
def potencia(base, exponente=2):
    return base ** exponente

print(potencia(5))        # usa exponente = 2
print(potencia(5, 3))     # usa exponente = 3
print(potencia(exponente=10, base=2))   # también se pueden nombrar los argumentos
````

## 5. Funciones que trabajan con listas

````python
def promedio(numeros):
    """Devuelve el promedio de una lista de números."""
    return sum(numeros) / len(numeros)

def aprobados(notas, minimo=3.0):
    """Devuelve la lista de notas mayores o iguales al mínimo."""
    return [n for n in notas if n >= minimo]

notas = [3.5, 4.2, 2.8, 4.9, 3.1, 4.0]
print("Promedio:", promedio(notas))
print("Aprobadas:", aprobados(notas))
print("Con mínimo 4.0:", aprobados(notas, 4.0))
print("Promedio de las aprobadas:", promedio(aprobados(notas)))
````

## 6. Las variables de una función son locales

Las variables que se crean **dentro** de una función solo existen mientras la función se ejecuta. Son **locales**: afuera no se ven.

````python
def calcular():
    resultado_secreto = 42
    return resultado_secreto

print(calcular())
print(resultado_secreto)      # ¿existe afuera?
````

`NameError`: `resultado_secreto` solo vivía dentro de `calcular`. Esto es bueno: cada función tiene su propio espacio y no "ensucia" las variables del resto del programa. Para sacar un valor de una función, se usa `return`.

## 7. 🎯 Los números amigos

Ahora sí. Escribimos **una vez** la función que suma los divisores propios…

````python
def suma_divisores(n):
    """Suma los divisores propios de n (todos menos n)."""
    total = 0
    for d in range(1, n):
        if n % d == 0:
            total += d
    return total

print("Divisores propios de 220 suman:", suma_divisores(220))
print("Divisores propios de 284 suman:", suma_divisores(284))
````

…y la usamos para crear **otra** función. Las funciones se pueden construir con otras funciones, como piezas de Lego.

````python
def son_amigos(a, b):
    return a != b and suma_divisores(a) == b and suma_divisores(b) == a

print(son_amigos(220, 284))
print(son_amigos(220, 285))
````

````python
# Buscar parejas de amigos menores que 1500
for a in range(2, 1500):
    b = suma_divisores(a)
    if a < b < 1500 and suma_divisores(b) == a:
        print("¡Amigos!", a, "y", b)
````

### Perfectos, abundantes y deficientes

Con la misma función clasificamos todos los números:
- **perfecto** si sus divisores propios suman exactamente el número (como 6 = 1 + 2 + 3);
- **abundante** si suman más;
- **deficiente** si suman menos.

````python
import matplotlib.pyplot as plt

def clasificar(n):
    s = suma_divisores(n)
    if s == n:
        return "perfecto"
    elif s > n:
        return "abundante"
    else:
        return "deficiente"

print([n for n in range(2, 1000) if clasificar(n) == "perfecto"])

numeros = list(range(2, 301))
sumas = [suma_divisores(n) for n in numeros]
colores = ["red" if s > n else "tab:blue" for n, s in zip(numeros, sumas)]

plt.figure(figsize=(9, 4.5))
plt.scatter(numeros, sumas, c=colores, s=12)
plt.plot(numeros, numeros, color="gray", linestyle="--", label="suma = n")
plt.xlabel("n")
plt.ylabel("suma de divisores propios")
plt.title("Rojos: abundantes · Azules: deficientes · En la línea: perfectos")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````
