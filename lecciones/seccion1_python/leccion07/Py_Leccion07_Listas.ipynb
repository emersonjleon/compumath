## 1. Crear una lista

Una **lista** guarda varios valores en orden, entre **corchetes** `[ ]` y separados por comas. Puede tener números, textos, booleanos… ¡o mezclas!

````python
notas = [3.5, 4.2, 2.8, 4.9, 3.1]
frutas = ["mango", "lulo", "guayaba", "maracuyá"]
vacia = []

print(notas)
print(frutas)
print("¿Cuántas frutas hay?", len(frutas))
print("Lista vacía:", vacia, "con", len(vacia), "elementos")
````

### Índices y rebanadas: igual que en los textos

Las listas se numeran **desde 0**, y funcionan los índices negativos y las rebanadas de la Lección 3.

```
 frutas =  "mango"  "lulo"  "guayaba"  "maracuyá"
 índice:      0        1        2          3
 negativo:   -4       -3       -2         -1
```

````python
print(frutas[0])       # el primero
print(frutas[-1])      # el último
print(frutas[1:3])     # posiciones 1 y 2 (el 3 no se incluye)
print(frutas[::-1])    # al revés
````

## 2. Las listas sí se pueden modificar

Recuerda que los textos son **inmutables**: `texto[0] = "P"` daba error. Las listas, en cambio, son **mutables**: se pueden cambiar por dentro.

````python
frutas = ["mango", "lulo", "guayaba", "maracuyá"]
frutas[1] = "pitaya"           # reemplaza el elemento de la posición 1
print(frutas)
````

### Agregar y quitar elementos

| Método | Qué hace |
|---|---|
| `lista.append(x)` | agrega `x` al **final** |
| `lista.insert(i, x)` | inserta `x` en la posición `i` |
| `lista.remove(x)` | quita la **primera** aparición de `x` |
| `lista.pop()` | quita el **último** y lo devuelve |
| `lista.pop(i)` | quita el de la posición `i` y lo devuelve |

````python
mochila = ["cuaderno", "lápiz"]
mochila.append("calculadora")
print("Después de append:", mochila)

mochila.insert(0, "agenda")
print("Después de insert:", mochila)

mochila.remove("lápiz")
print("Después de remove:", mochila)

sacado = mochila.pop()
print("Saqué:", sacado, "→ quedó:", mochila)
````

````python
# 'in' pregunta si un elemento está en la lista (como en los textos)
print("cuaderno" in mochila)
print("borrador" in mochila)
````

## 3. Recorrer y resumir una lista

Un `for` recorre los elementos **uno por uno**, igual que recorría las letras de un texto.

````python
notas = [3.5, 4.2, 2.8, 4.9, 3.1, 4.0]

for nota in notas:
    if nota >= 3.0:
        print(nota, "aprobó")
    else:
        print(nota, "reprobó")
````

Python trae funciones listas para resumir una lista de números:

````python
print("Cantidad:", len(notas))
print("Suma:    ", sum(notas))
print("Promedio:", sum(notas) / len(notas))
print("Mayor:   ", max(notas))
print("Menor:   ", min(notas))
print("Ordenada:", sorted(notas))
print("De mayor a menor:", sorted(notas, reverse=True))
````

⚠️ `sorted(lista)` crea una lista **nueva** ordenada y deja la original igual. `lista.sort()` ordena la **misma** lista y no devuelve nada.

````python
numeros = [5, 2, 9, 1]
ordenados = sorted(numeros)
print("sorted:", ordenados, " original:", numeros)

numeros.sort()
print("después de sort():", numeros)
````

### Posición y valor a la vez: `enumerate`

````python
podio = ["Ana", "Luis", "Sara"]
for posicion, nombre in enumerate(podio, start=1):
    print(f"Puesto {posicion}: {nombre}")
````

## 4. Construir listas en un ciclo

Un patrón muy común: empezar con una lista **vacía** y ir agregando con `append`. Es un acumulador, pero en vez de sumar, **guarda** cada valor.

````python
cuadrados = []
for n in range(1, 11):
    cuadrados.append(n ** 2)
print(cuadrados)
````

````python
# range también se convierte en lista directamente
print(list(range(0, 30, 3)))
````

### Comprensiones de lista: lo mismo en una línea

Python tiene una forma corta de construir listas, llamada **comprensión de lista**:

```python
[expresión for variable in secuencia]
[expresión for variable in secuencia if condición]
```

````python
cuadrados = [n ** 2 for n in range(1, 11)]
print(cuadrados)

pares = [n for n in range(1, 21) if n % 2 == 0]
print(pares)

frutas = ["mango", "lulo", "guayaba"]
en_mayusculas = [f.upper() for f in frutas]
print(en_mayusculas)
````

### De texto a lista y de lista a texto

`split()` parte un texto en una lista de palabras, y `" ".join(lista)` las vuelve a unir.

````python
frase = "el que lee mucho y anda mucho ve mucho y sabe mucho"
palabras = frase.split()
print(palabras)
print("Número de palabras:", len(palabras))
print("Veces que aparece 'mucho':", palabras.count("mucho"))
print("-".join(palabras[:4]))
````

## 5. 🎲 Lanzar dados miles de veces

Primero un solo dado: lanzamos 6000 veces y guardamos cada resultado en una lista.

````python
import random

lanzamientos = []
for i in range(6000):
    lanzamientos.append(random.randint(1, 6))

print("Primeros 20 lanzamientos:", lanzamientos[:20])
for cara in range(1, 7):
    print(f"La cara {cara} salió {lanzamientos.count(cara)} veces")
````

Cada cara sale **alrededor de 1000 veces**, porque todas tienen la misma probabilidad (1/6). Ahora, dos dados. Usamos una lista `conteos` donde la posición `s` guarda cuántas veces salió la suma `s`.

````python
import matplotlib.pyplot as plt

conteos = [0] * 13          # 13 ceros: posiciones 0 a 12 (usaremos de la 2 a la 12)

for i in range(10000):
    suma = random.randint(1, 6) + random.randint(1, 6)
    conteos[suma] += 1

sumas = list(range(2, 13))
veces = conteos[2:]          # quitamos las posiciones 0 y 1, que nunca se usan
for s, v in zip(sumas, veces):
    print(f"Suma {s:2d}: {v:5d} veces  " + "█" * (v // 50))

plt.bar(sumas, veces, color="#5b8def")
plt.xticks(sumas)
plt.xlabel("suma de los dos dados")
plt.ylabel("veces en 10 000 lanzamientos")
plt.title("¿Qué suma sale más?")
plt.show()
````

### 🎯 ¿Por qué gana el 7?

Hay **36** combinaciones posibles de dos dados (6 × 6). Construyamos la lista de todas las sumas:

````python
todas = [a + b for a in range(1, 7) for b in range(1, 7)]
print("Total de combinaciones:", len(todas))
for s in range(2, 13):
    print(f"Suma {s:2d}: {todas.count(s)} de 36 formas")
````

El 7 se puede formar de **6 formas** (1+6, 2+5, 3+4, 4+3, 5+2, 6+1), más que cualquier otra suma. El 2 y el 12 solo de **una** forma. Por eso, con muchos lanzamientos, el 7 sale cerca de 10 000 × 6/36 ≈ 1667 veces y el 2 cerca de 278.

## 6. Errores y trampas comunes

````python
# Error 1: índice fuera de rango
frutas = ["mango", "lulo", "guayaba"]
print(frutas[3])
````

````python
# Error 2: quitar algo que no está
frutas = ["mango", "lulo", "guayaba"]
frutas.remove("fresa")
````

**Trampa: dos nombres para la misma lista.** Al escribir `b = a`, `b` **no** es una copia: es otro nombre para la **misma** lista. Si cambias una, cambia la otra. Para copiar de verdad se usa `a.copy()`.

````python
a = [1, 2, 3]
b = a            # ¡no es una copia!
b.append(4)
print("a:", a)   # a también cambió

c = a.copy()     # esta sí es una copia
c.append(5)
print("a:", a, "  c:", c)
````
