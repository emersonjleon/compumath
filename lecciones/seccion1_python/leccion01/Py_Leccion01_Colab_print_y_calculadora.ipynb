## 2. Python como calculadora

Si escribes una operación **sin comillas**, Python la calcula.

````python
print(7 + 5)     # suma
print(7 - 5)     # resta
print(7 * 5)     # multiplicación (se usa *, no x)
print(7 / 5)     # división
````

Fíjate en la diferencia entre texto y número:

````python
print("7 + 5")   # con comillas: es un texto, se muestra tal cual
print(7 + 5)     # sin comillas: es una operación, se calcula
````

### Operadores especiales

| Operador | Qué hace | Ejemplo | Resultado |
|---|---|---|---|
| `**` | potencia | `2 ** 10` | 1024 |
| `//` | división entera (cociente) | `17 // 5` | 3 |
| `%` | residuo de la división | `17 % 5` | 2 |

Recuerda la división de la primaria: 17 ÷ 5 da **3** y sobran **2**.

````python
print(2 ** 10)    # 2 multiplicado por sí mismo 10 veces
print(17 // 5)    # ¿cuántas veces cabe 5 en 17?
print(17 % 5)     # ¿cuánto sobra?
````

````python
# Python no tiene problema con números gigantes:
print(2 ** 100)
````

### El orden de las operaciones

Python sigue las mismas reglas que en matemáticas: primero potencias, luego multiplicaciones y divisiones, y al final sumas y restas. Los **paréntesis** cambian el orden.

````python
print(2 + 3 * 4)      # primero 3 * 4
print((2 + 3) * 4)    # primero el paréntesis
print(2 * 3 ** 2)     # primero la potencia
````

## 3. Los errores son tus amigos

Todo el que programa se equivoca muchas veces al día. Lo importante es **leer el mensaje de error**: casi siempre dice qué pasó y en qué línea.

Ejecuta cada celda y lee la **última línea roja**.

````python
# Error 1: olvidamos cerrar las comillas
print("Hola)
````

`SyntaxError` significa que la instrucción **está mal escrita**: Python no entiende la "gramática". Aquí falta la comilla de cierre.

````python
# Error 2: escribimos mal el nombre de la función
Print("Hola")
````

`NameError: name 'Print' is not defined` significa que Python **no conoce ese nombre**. Python distingue mayúsculas de minúsculas: `print` funciona, `Print` no.

````python
# Error 3: dividir entre cero
print(10 / 0)
````

`ZeroDivisionError`: en matemáticas no se puede dividir entre cero, y en Python tampoco.

💡 **Consejo:** cuando veas un error, lee la última línea, mira la línea que señala la flecha y corrige. Luego vuelve a ejecutar.

## 4. Una sorpresa: Python también dibuja

No te preocupes por entender este código todavía (lo aprenderás en la lección 10). Solo ejecútalo y observa qué tan rápido crecen las potencias de 2 comparadas con los múltiplos de 2.

````python
import matplotlib.pyplot as plt

n = list(range(0, 11))
plt.plot(n, [2 * k for k in n], 'o-', label="2 × n  (multiplicar)")
plt.plot(n, [2 ** k for k in n], 's-', label="2 ** n  (potencia)")
plt.xlabel("n")
plt.ylabel("resultado")
plt.title("Multiplicar contra elevar a una potencia")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````
