## 1. Diccionarios: buscar por nombre

En una lista, los elementos se buscan por **posición** (0, 1, 2…). En un **diccionario** se buscan por una **clave**, que suele ser un texto.

Un diccionario se escribe entre **llaves** `{ }` con pares `clave: valor` separados por comas.

````python
estudiante = {
    "nombre": "Ana",
    "edad": 14,
    "nota": 4.2,
    "deporte": "fútbol",
}

print(estudiante)
print(estudiante["nombre"])       # se busca por la clave, con corchetes
print(estudiante["nota"])
print("Tiene", len(estudiante), "datos")
````

Compara con una lista: `["Ana", 14, 4.2, "fútbol"]` tiene los mismos datos, pero ¿cómo sabrías que la posición 2 es la nota? Con un diccionario el código **se explica solo**.

## 2. Agregar, cambiar y borrar

````python
estudiante["nota"] = 4.6            # la clave ya existe: se CAMBIA el valor
estudiante["grado"] = 9             # la clave no existe: se AGREGA
del estudiante["deporte"]           # se BORRA
print(estudiante)

print("¿Tiene 'edad'?", "edad" in estudiante)        # 'in' revisa las claves
print("¿Tiene 'correo'?", "correo" in estudiante)
````

````python
# Error: buscar una clave que no existe
print(estudiante["correo"])
````

`KeyError` significa que la clave **no está** en el diccionario. Para evitarlo, `get(clave, valor_por_defecto)` devuelve un valor de respaldo si la clave no existe:

````python
print(estudiante.get("correo", "sin correo"))
print(estudiante.get("nombre", "sin nombre"))
````

## 3. Recorrer un diccionario

| Método | Devuelve |
|---|---|
| `dic.keys()` | las claves |
| `dic.values()` | los valores |
| `dic.items()` | los pares (clave, valor) |

````python
precios = {"empanada": 2500, "arepa": 3000, "jugo": 4000, "buñuelo": 1500}

for producto in precios:                  # recorrer un diccionario da sus claves
    print(producto)

print("Total de la lista de precios:", sum(precios.values()))

for producto, precio in precios.items():
    print(f"{producto:10s} ${precio}")
````

````python
# Un pedido: cuántas unidades de cada producto
pedido = {"empanada": 4, "jugo": 2}

total = 0
for producto, cantidad in pedido.items():
    total += precios[producto] * cantidad
print("Total del pedido: $", total)
````

## 4. Contar con un diccionario

Este es uno de los usos más importantes. Para contar las letras de un texto:
- si la letra **no está** todavía en el diccionario, empieza en 1;
- si **ya está**, se le suma 1.

Con `get` las dos cosas caben en una línea: `conteo[letra] = conteo.get(letra, 0) + 1`.

````python
texto = "abracadabra"
conteo = {}
for letra in texto:
    conteo[letra] = conteo.get(letra, 0) + 1
print(conteo)
````

### Ordenar por cantidad

`sorted(conteo, key=conteo.get, reverse=True)` ordena las **claves** según su **valor**, de mayor a menor. (Le pasamos la función `conteo.get` para que la use como criterio: ¡las funciones también se pueden pasar como argumentos!)

````python
ordenadas = sorted(conteo, key=conteo.get, reverse=True)
for letra in ordenadas:
    print(letra, "→", conteo[letra])
````

## 5. Tuplas: listas que no cambian

Una **tupla** es como una lista, pero **inmutable**: una vez creada no se puede modificar. Se escribe con **paréntesis**.

Sirve para datos que van juntos y no deberían cambiar, como las coordenadas de un punto o una fecha.

````python
punto = (3, 4)
fecha = (20, 7, 1810)          # día, mes, año

print(punto[0], punto[1])      # índices, igual que en las listas
print("Año:", fecha[2])
print(len(fecha))
````

````python
# Error: las tuplas no se pueden modificar
punto = (3, 4)
punto[0] = 10
````

### Desempacar una tupla

Se pueden sacar todos los valores de una tupla en varias variables a la vez:

````python
punto = (3, 4)
x, y = punto
print("x =", x, " y =", y)
print("Distancia al origen:", (x ** 2 + y ** 2) ** 0.5)

# ¿Recuerdas el intercambio de vasos de la Lección 2? Con tuplas es una línea:
a, b = "jugo", "leche"
a, b = b, a
print(a, b)
````

### Devolver varios valores de una función

Una función puede devolver una tupla, y así entregar **varios resultados**:

````python
def division_completa(a, b):
    return a // b, a % b         # devuelve una tupla (cociente, residuo)

cociente, residuo = division_completa(100, 7)
print(f"100 = 7 × {cociente} + {residuo}")
````

### Tuplas como claves

Las claves de un diccionario tienen que ser inmutables, así que una tupla puede ser clave (una lista no). Esto sirve para guardar cosas en un **tablero**:

````python
tablero = {(0, 0): "🏠", (2, 3): "💎", (4, 1): "🐉"}

for fila in range(5):
    linea = ""
    for col in range(5):
        linea += tablero.get((fila, col), "⬜")
    print(linea)

print("En (2, 3) hay:", tablero[(2, 3)])
````

## 6. Listas de diccionarios: una tabla

Una lista donde cada elemento es un diccionario es como una **tabla**: cada diccionario es una fila. (¡Así se organizan los datos que usaremos en machine learning!)

````python
curso = [
    {"nombre": "Ana",  "nota": 4.2, "grado": 9},
    {"nombre": "Luis", "nota": 3.1, "grado": 10},
    {"nombre": "Sara", "nota": 4.8, "grado": 9},
    {"nombre": "Juan", "nota": 2.7, "grado": 10},
]

notas = [e["nota"] for e in curso]
print("Promedio del curso:", sum(notas) / len(notas))

mejor = curso[0]
for e in curso:
    if e["nota"] > mejor["nota"]:
        mejor = e
print("Mejor estudiante:", mejor["nombre"], "con", mejor["nota"])

print("Grado 9:", [e["nombre"] for e in curso if e["grado"] == 9])
````

## 7. 🎯 Descifrar el mensaje

Volvamos al mensaje interceptado. Paso 1: contar sus letras.

````python
import matplotlib.pyplot as plt

secreto = "ox ex veqkb no vk wkxmrk no meiy xywlbo xy aesoby kmybnkbwo xy rk wemry dsowzy aeo fsfsk ex rsnkvqy no vyc no vkxjk ox kcdsvvoby knkbqk kxdsqek bymsx pvkmy i qkvqy mybbonyb exk yvvk no kvqy wkc fkmk aeo mkbxoby ckvzsmyx vkc wkc xymroc neovyc i aeolbkxdyc vyc cklknyc voxdotkc vyc fsobxoc kvqex zkvywsxy no kxknsnebk vyc nywsxqyc myxcewskx vkc dboc zkbdoc no ce rkmsoxnk"

conteo = {}
for letra in secreto:
    if letra.isalpha():                     # solo letras, sin espacios
        conteo[letra] = conteo.get(letra, 0) + 1

ordenadas = sorted(conteo, key=conteo.get, reverse=True)
print("Las 5 letras más frecuentes:", [(l, conteo[l]) for l in ordenadas[:5]])

letras = sorted(conteo)
plt.figure(figsize=(10, 3.5))
plt.bar(letras, [conteo[l] for l in letras], color="#c4a7e7")
plt.title("Frecuencia de las letras en el mensaje cifrado")
plt.show()
````

En español, las dos letras más usadas son la **e** y la **a** (cada una es más del 10 % de las letras de un texto). Así que la letra más frecuente del mensaje cifrado probablemente es una **e** o una **a** disfrazada.

Escribamos funciones para calcular el desplazamiento y descifrar, y probemos las dos hipótesis:

````python
def desplazamiento(cifrada, original):
    """Cuántos lugares se movió la letra original para convertirse en la cifrada."""
    return (ord(cifrada) - ord(original)) % 26

def descifrar(texto, d):
    resultado = ""
    for c in texto:
        if c.isalpha():
            resultado += chr((ord(c) - 97 - d) % 26 + 97)
        else:
            resultado += c
    return resultado

mas_comun = ordenadas[0]
for hipotesis in ["e", "a"]:
    d = desplazamiento(mas_comun, hipotesis)
    print(f"Si '{mas_comun}' es '{hipotesis}', el desplazamiento es {d}:")
    print("   ", descifrar(secreto, d)[:70], "...")
````

¡Una de las dos hipótesis da un texto que se puede leer! Es el comienzo del **Quijote**. Descifremos el mensaje completo:

````python
d = desplazamiento(mas_comun, "a")
print("Desplazamiento:", d)
print(descifrar(secreto, d))
````

Así se rompió la cifra de César: con unas pocas líneas, un diccionario y un poco de estadística. Por eso hoy los sistemas de cifrado son mucho más complejos.
