## 1. Repaso: qué es un texto

Un texto (`str`) es una secuencia de caracteres entre comillas. Puedes usar comillas dobles o simples.

````python
saludo = "Hola"
nombre = 'Valentina'
print(saludo, nombre)
print(type(nombre))
````

````python
# len() cuenta cuántos caracteres tiene un texto (los espacios también cuentan)
print(len("Hola"))
print(len("Hola mundo"))
print(len(""))          # el texto vacío tiene 0 caracteres
````

### Caracteres especiales

Dentro de un texto, la barra invertida `\` tiene un significado especial:

| Escribes | Significa |
|---|---|
| `\n` | salto de línea |
| `\t` | tabulación (espacio grande) |
| `\"` | una comilla dentro del texto |

````python
print("Primera línea\nSegunda línea")
print("Nombre:\tAna")
print("Ella dijo: \"¡Hola!\"")
print('También puedo usar "comillas dobles" dentro de comillas simples')
````

````python
# Las comillas triples permiten escribir varias líneas seguidas
poema = """Caminante, no hay camino,
se hace camino al andar."""
print(poema)
````

## 2. Armar frases con f-strings

En la Lección 2 pegábamos textos con `+` y `str()`. Funciona, pero se enreda:

````python
nombre = "Ana"
edad = 14
print("Me llamo " + nombre + " y tengo " + str(edad) + " años.")
````

Una **f-string** es un texto con una `f` antes de las comillas. Todo lo que pongas entre llaves `{ }` se reemplaza por su valor. ¡Es mucho más fácil!

````python
nombre = "Ana"
edad = 14
print(f"Me llamo {nombre} y tengo {edad} años.")
print(f"En 5 años tendré {edad + 5} años.")    # también se pueden hacer cuentas adentro
````

````python
# Con :.2f se muestran exactamente 2 decimales
precio = 12500 / 3
print(precio)
print(f"Cada uno paga ${precio:.2f}")
print(f"Redondeado a entero: ${precio:.0f}")
````

````python
# Error común: olvidar la f. Las llaves salen tal cual.
nombre = "Ana"
print("Hola {nombre}")
````

## 3. Índices: sacar una letra

Cada carácter de un texto tiene una **posición** llamada **índice**. ¡Ojo! En Python se empieza a contar desde **0**.

```
 texto =   P   y   t   h   o   n
 índice:   0   1   2   3   4   5
 negativo:-6  -5  -4  -3  -2  -1
```

Los índices negativos cuentan desde el final: `-1` es la última letra.

````python
texto = "Python"
print(texto[0])      # primera letra
print(texto[3])      # cuarta letra
print(texto[-1])     # última letra
print(texto[-2])     # penúltima
````

````python
# Error: pedir una posición que no existe
texto = "Python"      # tiene 6 letras: índices del 0 al 5
print(texto[6])
````

`IndexError: string index out of range` significa que el índice está **fuera del rango** del texto.

````python
# Error: los textos no se pueden modificar letra por letra
texto = "Gato"
texto[0] = "P"
````

Los textos son **inmutables**: no se pueden cambiar por dentro. Lo que sí puedes hacer es crear un texto nuevo:

````python
texto = "Gato"
texto = "P" + texto[1:]     # "P" + "ato"
print(texto)
````

## 4. Rebanadas: sacar un pedazo

`texto[inicio:fin]` saca el pedazo desde `inicio` **hasta antes de** `fin`.

- Si omites `inicio`, empieza desde el principio.
- Si omites `fin`, llega hasta el final.
- `texto[::-1]` recorre el texto **al revés**.

````python
palabra = "murciélago"
print(palabra[0:4])     # posiciones 0, 1, 2, 3 → "murc"
print(palabra[:3])      # las 3 primeras
print(palabra[4:])      # desde la posición 4 hasta el final
print(palabra[-4:])     # las 4 últimas
print(palabra[::-1])    # al revés
````

### 🎯 Volvamos a los palíndromos

Una palabra es palíndromo si es **igual** a sí misma al revés. Recuerda que `==` compara y da `True` o `False`.

````python
palabra = "reconocer"
print(palabra == palabra[::-1])
````

````python
palabra = "python"
print(palabra == palabra[::-1])
````

¡Una sola línea! Pero con frases no funciona tan fácil:

````python
frase = "Anita lava la tina"
print(frase[::-1])
print(frase == frase[::-1])
````

Falla por dos razones: la `A` mayúscula no es igual a la `a` minúscula, y los espacios quedan en otros lugares. Para arreglarlo necesitamos los **métodos**.

## 5. Métodos: herramientas de los textos

Un **método** es una función que pertenece a un tipo de dato. Se usa con un punto: `texto.metodo()`. Los métodos **no cambian** el texto original; devuelven uno nuevo.

| Método | Qué hace | Ejemplo | Resultado |
|---|---|---|---|
| `.upper()` | todo en mayúsculas | `"hola".upper()` | `"HOLA"` |
| `.lower()` | todo en minúsculas | `"HoLa".lower()` | `"hola"` |
| `.title()` | Mayúscula Inicial | `"ana pérez".title()` | `"Ana Pérez"` |
| `.replace(a, b)` | cambia `a` por `b` | `"casa".replace("c", "p")` | `"pasa"` |
| `.count(x)` | cuántas veces aparece `x` | `"banana".count("a")` | `3` |
| `.find(x)` | posición donde aparece `x` (-1 si no está) | `"banana".find("n")` | `2` |
| `.strip()` | quita espacios de los extremos | `"  hola  ".strip()` | `"hola"` |

````python
frase = "  Me gusta programar en Python  "
print(frase.upper())
print(frase.lower())
print(frase.strip())
print(frase.replace("Python", "Colab"))
print(frase.count("a"))
print(frase.find("Python"))
````

````python
# El texto original NO cambia: hay que guardar el resultado
nombre = "ana"
nombre.upper()
print(nombre)            # sigue en minúsculas

nombre = nombre.upper()  # ahora sí lo guardamos
print(nombre)
````

````python
# Se pueden encadenar varios métodos, uno tras otro
frase = "Anita lava la tina"
limpia = frase.lower().replace(" ", "")
print(limpia)
print(limpia == limpia[::-1])     # ¡ahora sí es palíndromo!
````

### La palabra `in`

`in` pregunta si un texto está **dentro** de otro, y responde `True` o `False`.

````python
correo = "ana.perez@colegio.edu.co"
print("@" in correo)
print("gmail" in correo)
print(correo[correo.find("@") + 1:])     # todo lo que va después de la @
````

## 6. ¿Qué vocal se usa más?

En español, ¿cuál vocal aparece más? Pega un texto largo (una canción, un párrafo de un libro) y ejecuta la celda. No te preocupes por el código de la gráfica; aprenderás `matplotlib` en la Lección 10.

````python
import matplotlib.pyplot as plt

texto = """En un lugar de la Mancha, de cuyo nombre no quiero acordarme,
no ha mucho tiempo que vivía un hidalgo de los de lanza en astillero,
adarga antigua, rocín flaco y galgo corredor."""

t = texto.lower()
conteos = [t.count("a") + t.count("á"),
           t.count("e") + t.count("é"),
           t.count("i") + t.count("í"),
           t.count("o") + t.count("ó"),
           t.count("u") + t.count("ú")]
print("a, e, i, o, u:", conteos)

plt.bar(["a", "e", "i", "o", "u"], conteos, color="#5b8def")
plt.title("¿Cuántas veces aparece cada vocal?")
plt.ylabel("veces")
plt.show()
````
