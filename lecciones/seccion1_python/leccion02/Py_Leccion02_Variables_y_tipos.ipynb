## 1. Una variable es una caja con etiqueta

Una **variable** es un nombre que guarda un valor. Imagínala como una caja con una etiqueta: la etiqueta es el nombre y lo que hay adentro es el valor.

En Python, el signo `=` **no** significa "es igual a". Significa **"guarda en"**: calcula lo que está a la derecha y guárdalo en la caja de la izquierda.

```python
edad = 14            # guarda 14 en la caja llamada edad
print(edad)          # muestra lo que hay en la caja
print(edad + 10)     # usa el valor para calcular
```

¿Recuerdas los segundos vividos de la Lección 1? Con variables el cálculo se lee mucho mejor, y solo hay que cambiar la edad en **un** lugar.

```python
edad = 14
dias = edad * 365
horas = dias * 24
segundos = horas * 60 * 60

print("Días vividos:", dias)
print("Horas vividas:", horas)
print("Segundos vividos:", segundos)
```

### Cambiar el valor de una variable

Si guardas un valor nuevo en la misma caja, el valor viejo **se reemplaza**.

```python
puntos = 10
print("Al empezar:", puntos)

puntos = 25           # el 10 se borra y queda 25
print("Después:", puntos)
```

### 🎯 La respuesta: x = x + 1

Python lee la línea **de derecha a izquierda**:
1. Calcula `x + 1` con el valor que `x` tiene ahora.
2. Guarda el resultado en `x`, reemplazando el valor anterior.

Es como decir: "saca lo que hay en la caja, súmale 1 y vuelve a guardarlo".

```python
x = 5
x = x + 1      # 5 + 1 = 6, se guarda en x
print(x)

x = x + 1      # 6 + 1 = 7
print(x)

x += 10        # forma corta de escribir x = x + 10
print(x)
```

También existen `-=`, `*=` y `/=`. Por ejemplo, `vidas -= 1` resta una vida en un videojuego.

```python
vidas = 3
vidas -= 1
print("Te quedan", vidas, "vidas")
```

## 3. Tipos de datos

No todos los valores son iguales. Python distingue varios **tipos**, y la función `type()` te dice de qué tipo es un valor.

| Tipo | Nombre | Ejemplos |
|---|---|---|
| `int` | número entero | `14`, `-3`, `0` |
| `float` | número decimal | `3.5`, `-0.25`, `2.0` |
| `str` | texto (*string*) | `"hola"`, `"14"` |
| `bool` | verdadero o falso | `True`, `False` |

```python
print(type(14))
print(type(3.5))
print(type("hola"))
print(type(True))
```

⚠️ En Python los decimales se escriben con **punto**, no con coma: `3.5`, no `3,5`.

### Los booleanos salen de las comparaciones

Cuando comparas dos valores, el resultado es `True` (verdadero) o `False` (falso). Los usaremos mucho en la lección de condicionales.

| Comparación | Significado |
|---|---|
| `==` | ¿son iguales? (dos signos `=`) |
| `!=` | ¿son distintos? |
| `<`, `>` | menor que, mayor que |
| `<=`, `>=` | menor o igual, mayor o igual |

```python
edad = 14
print(edad > 12)        # ¿edad es mayor que 12?
print(edad == 18)       # ¿edad es igual a 18?
print(edad != 18)       # ¿edad es distinta de 18?

# and significa "y": las dos condiciones deben ser verdaderas (lo veremos en la Lección 4)
es_adolescente = edad >= 13 and edad <= 17    # ¡un booleano también se puede guardar!
print("¿Es adolescente?", es_adolescente)
```

### El tipo importa: `"14"` no es lo mismo que `14`

```python
print(14 + 14)          # dos números: los suma
print("14" + "14")      # dos textos: los pega
print("14" * 3)         # texto por número: lo repite
```

```python
# Error: no se puede sumar un texto con un número
print("14" + 14)
```

`TypeError` significa que la operación **no tiene sentido para esos tipos**: Python no sabe si quieres sumar o pegar.

## 4. Convertir de un tipo a otro

| Función | Convierte a | Ejemplo | Resultado |
|---|---|---|---|
| `int()` | entero | `int("14")` | `14` |
| `float()` | decimal | `float("2.5")` | `2.5` |
| `str()` | texto | `str(14)` | `"14"` |

```python
texto = "14"
numero = int(texto)          # de texto a entero
print(numero + 1)

print(float("2.5") * 2)      # de texto a decimal
print("Tengo " + str(14) + " años")   # de número a texto, para pegarlo
```

```python
# int() con un decimal NO redondea: corta la parte decimal
print(int(3.9))
print(round(3.9))     # round() sí redondea
```

```python
# Error: no se puede convertir cualquier texto en número
int("catorce")
```

`ValueError`: el tipo es correcto (un texto), pero su **valor** no se puede convertir.

## 5. Pedir datos con `input()`

`input("pregunta")` muestra la pregunta, **espera** a que la persona escriba algo y presione Enter, y devuelve lo que escribió.

Ejecuta la celda y escribe tu nombre en el recuadro que aparece.

⚠️ **Importante:** `input()` **siempre** devuelve un texto (`str`), aunque escribas un número. Observa:

### Tu primer programa interactivo: convertir monedas

Este programa pide una cantidad de pesos colombianos y la convierte a dólares. Pruébalo varias veces con valores distintos.

```python
# Datos
tasa = 4000      # pesos por dólar (valor aproximado; cámbialo si quieres)

# Entrada
pesos = float(input("¿Cuántos pesos tienes? "))

# Proceso
dolares = pesos / tasa

# Salida
print(pesos, "pesos son", round(dolares, 2), "dólares")
```

💡 Casi todos los programas siguen esta estructura: **datos → entrada → proceso → salida**.

## 6. Mirar adentro de las cajas

Ejecuta esta celda para ver un dibujo de las variables que tiene guardadas un programa, con su nombre, su valor y su tipo. Aún no necesitas entender el código.

```python
import matplotlib.pyplot as plt

cajas = {"nombre": "Ana", "edad": 14, "estatura": 1.62, "es_estudiante": True}

fig, ax = plt.subplots(figsize=(10, 2.6))
colores = {"str": "#f6c177", "int": "#9ccfd8", "float": "#c4a7e7", "bool": "#a6da95"}
for i, (nom, val) in enumerate(cajas.items()):
    tipo = type(val).__name__
    ax.add_patch(plt.Rectangle((i * 2.5, 0), 2, 1.3, color=colores[tipo], ec="black"))
    ax.text(i * 2.5 + 1, 1.5, nom, ha="center", fontsize=12, weight="bold")
    ax.text(i * 2.5 + 1, 0.75, repr(val), ha="center", fontsize=13)
    ax.text(i * 2.5 + 1, 0.25, tipo, ha="center", fontsize=10, style="italic")
ax.set_xlim(-0.3, 10); ax.set_ylim(-0.2, 2)
ax.axis("off")
ax.set_title("Cada variable es una caja: etiqueta (nombre), contenido (valor) y tipo")
plt.show()
```
