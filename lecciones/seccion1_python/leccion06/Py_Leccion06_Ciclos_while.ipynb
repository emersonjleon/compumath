## 🔍 Doblar un papel hasta la Luna

Una hoja de papel tiene unos **0,1 milímetros** de grosor. Cada vez que la doblas por la mitad, su grosor se **duplica**.

La Luna está a unos **384 400 kilómetros** de la Tierra.

✏️ *Si pudieras doblar la hoja todas las veces que quisieras, ¿cuántos dobleces harían falta para que su grosor llegue a la Luna? ¿Miles? ¿Millones?*

**Mi predicción:** 

Esta vez **no sabemos** cuántas vueltas dará el ciclo: queremos repetir **hasta que** el grosor llegue a la Luna. Para eso está `while`.

## 1. `while`: mientras se cumpla la condición

```python
while condición:
    instrucciones
```

`while` significa **mientras**. Antes de cada vuelta, Python revisa la condición:
- si es `True`, ejecuta el bloque y vuelve a revisar;
- si es `False`, sale del ciclo y sigue con lo que viene después.

````python
n = 1
while n <= 5:
    print("n vale", n)
    n = n + 1          # ¡importante! cambiar la variable para que algún día la condición sea False
print("Terminé. Ahora n vale", n)
````

Este ciclo hace lo mismo que `for n in range(1, 6)`. Con `while` tenemos que hacer tres cosas a mano:
1. **Crear** la variable antes del ciclo (`n = 1`).
2. **Revisar** la condición (`n <= 5`).
3. **Actualizar** la variable dentro del ciclo (`n = n + 1`).

| Usa `for` cuando… | Usa `while` cuando… |
|---|---|
| sabes **cuántas** vueltas habrá | repites **hasta que** pase algo |
| recorres un `range` o un texto | no sabes de antemano cuántas vueltas serán |

## 2. 🎯 Volvamos a la Luna

````python
grosor = 0.1 / 1000 / 1000      # 0,1 mm en kilómetros
distancia_luna = 384400         # kilómetros
dobleces = 0

while grosor < distancia_luna:
    grosor = grosor * 2
    dobleces += 1

print(f"Se necesitan {dobleces} dobleces.")
print(f"El grosor sería de {grosor:.0f} km.")
````

¡Solo **42 dobleces**! Duplicar crece tan rápido que en 42 pasos pasamos de una décima de milímetro a más de 400 000 km. (En la vida real nadie ha logrado doblar un papel más de 12 veces).

Veamos cuántos dobleces hacen falta para otras distancias:

````python
grosor = 0.1 / 1000 / 1000
dobleces = 0

while grosor < 1:          # ¿cuándo llega a 1 km?
    grosor *= 2
    dobleces += 1
print("Para llegar a 1 km:", dobleces, "dobleces")

while grosor < 8.8:        # seguimos desde donde íbamos hasta la altura del Everest (8,8 km)
    grosor *= 2
    dobleces += 1
print("Para superar el Everest:", dobleces, "dobleces")
````

## 3. ⚠️ Ciclos infinitos

Si la condición **nunca** se vuelve `False`, el ciclo no termina nunca. Es un **ciclo infinito**. Por ejemplo, si olvidas actualizar la variable:

```python
n = 1
while n <= 5:
    print(n)       # n siempre vale 1... ¡para siempre!
```

**No ejecutes ese código.** Si alguna vez un ciclo infinito te ocurre en Colab, verás que el botón ▶️ se convierte en un cuadrado ⏹️ que gira. Haz clic en él (o usa el menú **Entorno de ejecución → Interrumpir ejecución**) para detenerlo.

💡 Un truco de seguridad mientras aprendes: agrega un **límite de vueltas** a la condición.

````python
x = 0
vueltas = 0
while x != 1 and vueltas < 20:     # el límite evita que el ciclo sea infinito
    x = x + 0.1
    vueltas += 1
print("x =", x, "después de", vueltas, "vueltas")
````

¿Sorprendido? Sumando 0.1 diez veces **no** da exactamente 1 (recuerda que `0.1 + 0.2` daba `0.30000000000000004`). Sin el límite, `x != 1` sería siempre verdadero y este ciclo nunca terminaría. Moraleja: con decimales es más seguro usar `<` o `>` que `!=`.

## 4. Insistir hasta que la respuesta sea válida

Un uso muy común de `while`: volver a preguntar **mientras** la respuesta no tenga sentido.

````python
nota = float(input("Escribe una nota entre 1.0 y 5.0: "))

while nota < 1.0 or nota > 5.0:
    print("Esa nota no es válida, inténtalo de nuevo.")
    nota = float(input("Escribe una nota entre 1.0 y 5.0: "))

print("Gracias, registré la nota", nota)
````

## 5. `break` y `continue`

- `break` **sale** del ciclo inmediatamente, aunque la condición siga siendo verdadera.
- `continue` **salta** el resto de esta vuelta y pasa a la siguiente.

Ambos funcionan en `while` y en `for`.

````python
# break: buscar el primer múltiplo de 7 mayor que 100
n = 101
while True:                # "mientras verdadero": un ciclo que solo termina con break
    if n % 7 == 0:
        break
    n += 1
print("El primer múltiplo de 7 mayor que 100 es", n)
````

````python
# continue: mostrar del 1 al 20, saltando los múltiplos de 3
for n in range(1, 21):
    if n % 3 == 0:
        continue           # salta el print de esta vuelta
    print(n, end=" ")
````

````python
# while True + break: repetir hasta que el usuario escriba "salir"
while True:
    palabra = input("Escribe una palabra (o 'salir'): ").strip().lower()
    if palabra == "salir":
        break
    print(f"'{palabra}' tiene {len(palabra)} letras")
print("¡Hasta luego!")
````

## 6. 🎮 Adivina el número

El computador piensa un número del 1 al 100 y te da pistas. El ciclo se repite **mientras** no aciertes, así que no sabemos cuántas vueltas dará.

````python
import random

secreto = random.randint(1, 100)
intentos = 0
intento = 0                   # cualquier valor distinto del secreto para empezar

print("Estoy pensando un número del 1 al 100...")

while intento != secreto:
    intento = int(input("Tu intento: "))
    intentos += 1
    if intento < secreto:
        print("  Más alto ⬆️")
    elif intento > secreto:
        print("  Más bajo ⬇️")

print(f"¡Correcto! Era {secreto}. Lo lograste en {intentos} intentos. 🎉")
````

💡 **Estrategia:** si siempre dices el número de la **mitad** del rango que queda (50, luego 25 o 75…), nunca necesitas más de **7 intentos**. ¿Por qué? Cada pista parte el rango por la mitad: 100 → 50 → 25 → 13 → 7 → 4 → 2 → 1.

## 7. La conjetura de Collatz

Toma un número entero positivo. Si es par, divídelo entre 2; si es impar, multiplícalo por 3 y súmale 1. Repite **hasta llegar a 1**.

Nadie ha demostrado que **siempre** se llega a 1, pero se ha comprobado con computadores para números gigantescos. ¡Un problema perfecto para `while`!

````python
import matplotlib.pyplot as plt

n = 27
pasos = 0
plt.scatter(pasos, n, color="tab:purple", s=12)

while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    pasos += 1
    plt.scatter(pasos, n, color="tab:purple", s=12)

print("El 27 llega a 1 en", pasos, "pasos")
plt.title("Collatz empezando en 27")
plt.xlabel("paso")
plt.ylabel("valor")
plt.grid(alpha=0.3)
plt.show()
````

El 27 sube hasta **9232** antes de caer a 1. Cambia el `27` por otros números y compara.

## 8. Errores comunes

````python
# Error: usar una variable en la condición sin haberla creado antes
while contador < 5:
    contador += 1
````

`NameError`: el `while` revisa la condición **antes** de la primera vuelta, así que `contador` tiene que existir desde antes.

````python
# Error de lógica: la condición es falsa desde el principio, así que el ciclo no da ninguna vuelta
n = 10
while n < 5:
    print("Nunca me verás")
    n += 1
print("El ciclo dio cero vueltas")
````
