## 1. `if`: hacer algo solo si se cumple una condición

```python
if condición:
    instrucciones que se ejecutan solo si la condición es True
```

Tres detalles muy importantes:
1. La línea del `if` termina con **dos puntos** `:`.
2. Las instrucciones de adentro van **corridas a la derecha** (4 espacios o la tecla Tab). A eso se le llama **indentación**.
3. La condición es cualquier cosa que dé `True` o `False`, como las comparaciones de la Lección 2.

````python
temperatura = 32

if temperatura > 30:
    print("¡Hace mucho calor!")       # está indentado: solo se ejecuta si la condición es True
    print("Toma agua.")

print("Esta línea siempre se ejecuta.")   # sin indentación: está fuera del if
````

## 2. `else`: si no, haz otra cosa

`else` (que significa *si no*) se ejecuta cuando la condición del `if` es `False`. No lleva condición, pero sí lleva dos puntos.

````python
edad = 15

if edad >= 18:
    print("Puedes votar en las elecciones.")
else:
    print("Todavía no puedes votar.")
    print(f"Te faltan {18 - edad} años.")
````

````python
# ¿Par o impar? Un número es par si el residuo de dividirlo entre 2 es 0
numero = 17

if numero % 2 == 0:
    print(f"{numero} es par")
else:
    print(f"{numero} es impar")
````

## 3. `elif`: varias opciones

Cuando hay más de dos caminos se usa `elif` (abreviatura de *else if*, "si no, si…"). Python revisa las condiciones **de arriba hacia abajo** y ejecuta **solo la primera** que sea verdadera.

Ejemplo: en Colombia las notas van de 1.0 a 5.0 y muchos colegios usan esta escala de desempeño.

| Nota | Desempeño |
|---|---|
| 4.6 a 5.0 | Superior |
| 4.0 a 4.5 | Alto |
| 3.0 a 3.9 | Básico |
| menos de 3.0 | Bajo |

````python
nota = 4.2

if nota >= 4.6:
    print("Desempeño superior 🌟")
elif nota >= 4.0:
    print("Desempeño alto 👍")
elif nota >= 3.0:
    print("Desempeño básico")
else:
    print("Desempeño bajo: ¡a estudiar!")
````

¿Por qué en el segundo caso basta con `nota >= 4.0`, sin decir también `nota < 4.6`? Porque si la nota fuera 4.6 o más, Python ya habría entrado en el primer `if` y **no revisaría** los demás.

⚠️ Por eso **el orden importa**. Mira qué pasa si ponemos las condiciones al revés:

````python
nota = 4.8

if nota >= 3.0:
    print("Desempeño básico")       # ¡4.8 >= 3.0 es True, así que entra aquí!
elif nota >= 4.0:
    print("Desempeño alto")
elif nota >= 4.6:
    print("Desempeño superior")
````

Una nota de 4.8 quedó como "básico". El programa no da error, pero está **mal**: es un error de lógica, el más difícil de encontrar.

## 4. Combinar condiciones: `and`, `or`, `not`

| Operador | Significa | Es `True` cuando… |
|---|---|---|
| `a and b` | a **y** b | las dos son verdaderas |
| `a or b` | a **o** b | al menos una es verdadera |
| `not a` | **no** a | a es falsa |

````python
print(True and True, True and False, False and False)
print(True or True, True or False, False or False)
print(not True, not False)
````

````python
edad = 15
tiene_permiso = True

if edad >= 14 and tiene_permiso:
    print("Puedes ir a la salida de campo.")

dia = "sábado"
if dia == "sábado" or dia == "domingo":
    print("¡Es fin de semana!")

lloviendo = False
if not lloviendo:
    print("Podemos jugar fútbol afuera.")
````

### Un ejemplo matemático: los años bisiestos

Un año es **bisiesto** (tiene 29 de febrero) si es divisible entre 4, **excepto** los que son divisibles entre 100, que solo son bisiestos si también son divisibles entre 400.

En una sola condición: *(divisible entre 4 **y no** entre 100) **o** divisible entre 400*.

````python
anio = 2024

if (anio % 4 == 0 and anio % 100 != 0) or anio % 400 == 0:
    print(f"{anio} es bisiesto")
else:
    print(f"{anio} no es bisiesto")
````

## 5. Decisiones dentro de decisiones

Un `if` puede ir dentro de otro. Cada nivel se indenta un poco más.

````python
numero = -8

if numero > 0:
    print("Positivo")
else:
    if numero == 0:
        print("Cero")
    else:
        print("Negativo")
        if numero % 2 == 0:
            print("…y además es par")
````

Muchas veces un `elif` hace lo mismo y se lee mejor. Ambas formas son correctas.

## 7. 🎮 Piedra, papel o tijera

El computador elige al azar con `random.randint(1, 3)`, que da un número entero entre 1 y 3. Lo convertimos en una jugada y luego decidimos quién gana.

Ejecuta la celda varias veces y escribe `piedra`, `papel` o `tijera`.

````python
import random

# El computador elige
numero = random.randint(1, 3)
if numero == 1:
    compu = "piedra"
elif numero == 2:
    compu = "papel"
else:
    compu = "tijera"

# Tú eliges (.strip().lower() limpia lo que escribas)
tu = input("¿Piedra, papel o tijera? ").strip().lower()
print(f"Tú: {tu}  ·  Computador: {compu}")

# El árbitro decide
if tu == compu:
    print("¡Empate! 🤝")
elif (tu == "piedra" and compu == "tijera") or \
     (tu == "tijera" and compu == "papel") or \
     (tu == "papel" and compu == "piedra"):
    print("¡Ganaste! 🎉")
elif tu == "piedra" or tu == "papel" or tu == "tijera":
    print("Ganó el computador 🤖")
else:
    print("Esa jugada no existe. Escribe piedra, papel o tijera.")
````

💡 La barra `\` al final de una línea le dice a Python que la instrucción **continúa** en la línea siguiente. Sirve para partir condiciones largas.

**Respuesta a la pregunta inicial:** hay 3 × 3 = **9** combinaciones: 3 empates, 3 en que ganas y 3 en que pierdes.
