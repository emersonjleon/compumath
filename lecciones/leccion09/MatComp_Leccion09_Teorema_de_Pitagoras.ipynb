## 📚 El concepto: el teorema de Pitágoras

En un **triángulo rectángulo**, los dos lados que forman el ángulo recto se llaman **catetos** (a y b), y el lado opuesto, el más largo, es la **hipotenusa** (c).

$$a^2 + b^2 = c^2$$

**Interpretación geométrica:** el área del cuadrado construido sobre la hipotenusa es igual a la suma de las áreas de los cuadrados construidos sobre los catetos.

Con 3, 4 y 5: 3² + 4² = 9 + 16 = 25 = 5². ✔

## 1. Calcular lados

```python
import math

def hipotenusa(a, b):
    return math.sqrt(a**2 + b**2)

def cateto(c, a):
    """El otro cateto, conociendo la hipotenusa c y un cateto a."""
    return math.sqrt(c**2 - a**2)

print("Catetos 3 y 4 → hipotenusa", hipotenusa(3, 4))
print("Catetos 1 y 1 → hipotenusa", hipotenusa(1, 1))
print("Escalera de 5 m a 3 m de la pared → llega a", cateto(5, 3), "m de altura")
```

## 2. Viendo el teorema: cuadrados sobre los lados

Dibujemos un triángulo rectángulo con un cuadrado sobre cada lado. Mueve los sliders para cambiar los catetos y compara las áreas.

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, FloatSlider

def cuadrado_sobre(ax, P, Q, R, color):
    """Dibuja el cuadrado sobre el lado PQ, hacia afuera del triángulo PQR."""
    P, Q, R = np.array(P, float), np.array(Q, float), np.array(R, float)
    d = Q - P
    normal = np.array([d[1], -d[0]])        # perpendicular a PQ
    if normal @ (R - P) > 0:                # si apunta hacia el triángulo, la volteamos
        normal = -normal
    vertices = [P, Q, Q + normal, P + normal]
    ax.add_patch(plt.Polygon(vertices, color=color, alpha=0.5))
    area = d @ d
    centro = (P + Q) / 2 + normal / 2
    ax.text(*centro, f"{area:.1f}", ha="center", va="center", fontsize=13)

def ver_cuadrados(a, b):
    A, B, Cc = (0, 0), (a, 0), (0, b)       # ángulo recto en A
    fig, ax = plt.subplots(figsize=(7, 7))
    cuadrado_sobre(ax, A, B, Cc, "#2a9d8f")      # cuadrado sobre a
    cuadrado_sobre(ax, A, Cc, B, "#e9c46a")     # cuadrado sobre b
    cuadrado_sobre(ax, B, Cc, A, "#e76f51")     # cuadrado sobre c
    ax.add_patch(plt.Polygon([A, B, Cc], color="#264653"))
    ax.set_xlim(-b - 1, a + b + 1); ax.set_ylim(-a - 1, a + b + 1)
    ax.set_aspect("equal"); ax.axis("off")
    c = hipotenusa(a, b)
    ax.set_title(f"a² + b² = {a**2:.1f} + {b**2:.1f} = {a**2 + b**2:.1f}     c² = {c**2:.1f}")
    plt.show()

interact(ver_cuadrados,
         a=FloatSlider(value=3, min=1, max=6, step=0.5, continuous_update=False),
         b=FloatSlider(value=4, min=1, max=6, step=0.5, continuous_update=False));
```

### Una demostración sin palabras

Toma un cuadrado grande de lado **a + b** y coloca dentro **4 copias** del triángulo de dos maneras distintas:

- **Izquierda:** el espacio que sobra es un cuadrado inclinado de lado **c** → área c².
- **Derecha:** el espacio que sobra son dos cuadrados, de lados **a** y **b** → área a² + b².

¡El cuadrado grande y los 4 triángulos son iguales en los dos dibujos, así que lo que sobra también tiene que ser igual!

```python
def demostracion(a, b):
    L = a + b
    fig, axes = plt.subplots(1, 2, figsize=(11, 5.5))
    tri = "#264653"
    # Izquierda: 4 triángulos en las esquinas, cuadrado c² inclinado en el centro
    ax = axes[0]
    ax.add_patch(plt.Polygon([(0, 0), (L, 0), (L, L), (0, L)], color="#e76f51", alpha=0.5))
    for t in [[(0, 0), (a, 0), (0, b)], [(a, 0), (L, 0), (L, a)],
              [(L, a), (L, L), (b, L)], [(b, L), (0, L), (0, b)]]:
        ax.add_patch(plt.Polygon(t, color=tri, ec="white"))
    ax.text(L / 2, L / 2, "c²", ha="center", va="center", fontsize=20)
    # Derecha: los 4 triángulos forman 2 rectángulos; quedan a² y b²
    ax = axes[1]
    ax.add_patch(plt.Polygon([(0, 0), (a, 0), (a, a), (0, a)], color="#2a9d8f", alpha=0.5))
    ax.add_patch(plt.Polygon([(a, a), (L, a), (L, L), (a, L)], color="#e9c46a", alpha=0.5))
    for t in [[(a, 0), (L, 0), (L, a)], [(a, 0), (a, a), (L, a)],
              [(0, a), (a, a), (0, L)], [(a, a), (a, L), (0, L)]]:
        ax.add_patch(plt.Polygon(t, color=tri, ec="white"))
    ax.text(a / 2, a / 2, "a²", ha="center", va="center", fontsize=20)
    ax.text(a + b / 2, a + b / 2, "b²", ha="center", va="center", fontsize=20)
    for ax in axes:
        ax.set_xlim(-0.2, L + 0.2); ax.set_ylim(-0.2, L + 0.2)
        ax.set_aspect("equal"); ax.axis("off")
    axes[0].set_title("lo que sobra: c²")
    axes[1].set_title("lo que sobra: a² + b²")
    plt.show()

interact(demostracion,
         a=FloatSlider(value=3, min=1, max=5, step=0.5, continuous_update=False),
         b=FloatSlider(value=4, min=1, max=5, step=0.5, continuous_update=False));
```

## 3. ¿Recto, agudo u obtuso?

El teorema también funciona **al revés**: si a² + b² = c², el triángulo es rectángulo. Y si no:

| Comparación (c es el lado mayor) | Tipo de triángulo |
|---|---|
| c² < a² + b² | **acutángulo** (todos los ángulos agudos) |
| c² = a² + b² | **rectángulo** |
| c² > a² + b² | **obtusángulo** (un ángulo mayor de 90°) |

```python
def clasificar(x, y, z):
    a, b, c = sorted([x, y, z])          # c es el mayor
    if a + b <= c:
        return "no forman un triángulo"
    if a**2 + b**2 == c**2:
        return "rectángulo"
    elif a**2 + b**2 > c**2:
        return "acutángulo"
    else:
        return "obtusángulo"

for lados in [(3, 4, 5), (6, 8, 10), (5, 12, 13), (2, 3, 4), (4, 5, 6), (1, 2, 3), (7, 24, 25)]:
    print(lados, "→", clasificar(*lados))
```

¡Ya podemos responder la pregunta del albañil! 6-8-10 y 5-12-13 también sirven; 2-3-4 da un ángulo **obtuso**.

## 4. Ternas pitagóricas

Una **terna pitagórica** son tres enteros positivos (a, b, c) con a² + b² = c². Busquemos todas con c ≤ 100.

```python
ternas = []
for c in range(1, 101):
    for a in range(1, c):
        for b in range(a, c):
            if a**2 + b**2 == c**2:
                ternas.append((a, b, c))

print("Hay", len(ternas), "ternas con c ≤ 100")
print(ternas[:15])
```

Algunas son "copias agrandadas" de otras: (6, 8, 10) = 2 · (3, 4, 5). Una terna es **primitiva** si sus números no tienen divisores comunes, es decir, si **MCD(a, b) = 1** (Lección 4).

```python
primitivas = [t for t in ternas if math.gcd(t[0], t[1]) == 1]
print("Primitivas:", len(primitivas))
for t in primitivas:
    print(t)
```

```python
# Dibujemos todas las parejas (a, b) de catetos enteros con c ≤ 300
puntos = []
for a in range(1, 300):
    for b in range(1, 300):
        c = math.isqrt(a * a + b * b)
        if c * c == a * a + b * b and c <= 300:
            puntos.append((a, b, math.gcd(a, b) == 1))

plt.figure(figsize=(7, 7))
plt.scatter([p[0] for p in puntos if not p[2]], [p[1] for p in puntos if not p[2]], s=4, color="lightgray", label="múltiplos")
plt.scatter([p[0] for p in puntos if p[2]], [p[1] for p in puntos if p[2]], s=8, color="crimson", label="primitivas")
plt.xlabel("a"); plt.ylabel("b"); plt.legend()
plt.gca().set_aspect("equal")
plt.title("Catetos enteros de triángulos rectángulos")
plt.show()
```

Los puntos forman **líneas rectas** desde el origen (los múltiplos de cada terna primitiva) y unas **curvas** como parábolas. ¿Hay alguna regla que genere todas las primitivas?

**Fórmula de Euclides:** si m > n > 0 son enteros, entonces
$$a = m^2 - n^2, \qquad b = 2mn, \qquad c = m^2 + n^2$$
siempre es una terna pitagórica.

```python
def euclides(m, n):
    return m**2 - n**2, 2 * m * n, m**2 + n**2

print(" m  n   terna            ¿cumple?")
for m in range(2, 7):
    for n in range(1, m):
        a, b, c = euclides(m, n)
        print(f"{m:2} {n:2}   {(a, b, c)!s:15}  {a**2 + b**2 == c**2}")
```

## 5. Distancias y la espiral de Teodoro

La distancia entre dos puntos (x₁, y₁) y (x₂, y₂) es la hipotenusa de un triángulo cuyos catetos son Δx y Δy:

$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

```python
def distancia(P, Q):
    return math.sqrt((Q[0] - P[0])**2 + (Q[1] - P[1])**2)

casa, colegio = (1, 2), (7, 10)
print("Distancia en línea recta:", distancia(casa, colegio))
print("Distancia caminando por cuadras (taxista):", abs(7 - 1) + abs(10 - 2))
```

**La espiral de Teodoro** (siglo V a. C.): empieza con un triángulo rectángulo de catetos 1 y 1, cuya hipotenusa es √2. Sobre esa hipotenusa pon otro triángulo con un cateto de 1: su hipotenusa es √3. Repite… las hipotenusas valen √2, √3, √4, √5, …

Mueve el slider para agregar triángulos.

```python
from ipywidgets import IntSlider

def teodoro(n):
    fig, ax = plt.subplots(figsize=(7, 7))
    angulo = 0.0
    colores = plt.cm.viridis(np.linspace(0, 1, n))
    for k in range(1, n + 1):
        r = math.sqrt(k)                      # cateto largo (hipotenusa anterior)
        P = (r * math.cos(angulo), r * math.sin(angulo))
        dtheta = math.atan(1 / r)             # ángulo que agrega el cateto de 1
        angulo2 = angulo + dtheta
        R = math.sqrt(k + 1)
        Q = (R * math.cos(angulo2), R * math.sin(angulo2))
        ax.add_patch(plt.Polygon([(0, 0), P, Q], color=colores[k - 1], ec="white", zorder=-k))   # los primeros quedan encima
        if k <= 16:
            ax.text(Q[0] * 1.08, Q[1] * 1.08, f"√{k + 1}", ha="center", va="center", fontsize=9, zorder=5)
        angulo = angulo2
    lim = math.sqrt(n + 1) + 0.6
    ax.set_xlim(-lim, lim); ax.set_ylim(-lim, lim)
    ax.set_aspect("equal"); ax.axis("off")
    ax.set_title(f"Espiral de Teodoro con {n} triángulos: la última hipotenusa mide √{n + 1} ≈ {math.sqrt(n + 1):.3f}")
    plt.show()

interact(teodoro, n=IntSlider(value=16, min=1, max=100, continuous_update=False));
```
