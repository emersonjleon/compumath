## 📚 El concepto: la computadora como laboratorio

Python trabaja con números enteros **de cualquier tamaño y de forma exacta**: puede calcular 2¹⁰⁰ o 52! sin redondear nada.

Nuestro método en todo el curso será:
1. **Experimentar**: la computadora produce muchos casos.
2. **Conjeturar**: nosotros proponemos una regla.
3. **Poner a prueba**: buscamos el caso que rompa la regla.

Una conjetura que funciona en muchos casos **no está demostrada**. ¡Hoy veremos un ejemplo!

## 2. Python como calculadora

| Operación | Símbolo | Ejemplo |
|---|---|---|
| Suma | `+` | `7 + 3` |

| Multiplicación | `*` | `7 * 3` |

````python
print(7 + 3)
print(7 - 3)
print(7 * 3)
print(7 / 3)
print(7 ** 3)
````

## 3. Cociente y residuo: `//` y `%`

Cuando dividimos 17 entre 5: el **cociente** es 3 y el **residuo** es 2, porque 17 = 5 × 3 + 2.

Trucos muy útiles:
- `n % 2 == 0` nos dice si n es **par**.
- `n % 10` es la **última cifra** de n.
- `n // 10` **quita** la última cifra.

````python
print(17 // 5, 17 % 5)
n = 2026
print("¿Es par?", n % 2 == 0)
print("Última cifra:", n % 10)
print("Sin la última cifra:", n // 10)
````

## 4. Variables: ponerle nombre a los números

Una **variable** guarda un valor para usarlo después. ¿Cuántos segundos has vivido?
Cambia `edad` por tu edad y ejecuta.

````python
edad = 14
segundos_por_año = 365 * 24 * 60 * 60
segundos_vividos = edad * segundos_por_año
print("Segundos en un año:", segundos_por_año)
print("Has vivido aproximadamente", segundos_vividos, "segundos")
````

## 5. Números gigantes

Python no se asusta con números enormes. Para contar sus cifras, lo convertimos en texto con `str` y medimos su longitud con `len`.

````python
import math

print(2 ** 100)
print("Cifras de 2^100:", len(str(2 ** 100)))
print("52! =", math.factorial(52))      # formas de ordenar una baraja
print("Cifras de 52!:", len(str(math.factorial(52))))
````

## 6. El ciclo `for`: repetir para encontrar patrones

`for k in range(1, 11):` repite el bloque de abajo con k = 1, 2, …, 10. ¡Ojo con la **sangría** (los espacios al inicio)!

````python
# La tabla del 7
for k in range(1, 11):
    print(7, "x", k, "=", 7 * k)
````

### Volvamos a los unos
Un número formado solo por unos se llama **repuno**. Elevemos al cuadrado los repunos de 1 a 9 unos.

````python
repuno = 0
for k in range(1, 10):
    repuno = repuno * 10 + 1       # 1, 11, 111, ...
    print(repuno, "^2 =", repuno ** 2)
````

✏️ **Conjetura:** ¿cómo será el cuadrado del repuno de 10, 11 y 12 unos? Escríbelo antes de ejecutar la celda siguiente.

````python
repuno = 0
for k in range(1, 13):
    repuno = repuno * 10 + 1
    if k >= 10:
        print(k, "unos:", repuno ** 2)
````

😮 ¿Se cumplió tu conjetura? El patrón **se rompe** por el **acarreo**: al sumar las columnas de la multiplicación aparece un 10 que "se lleva".
Esto es lo más importante de la lección: **muchos casos que funcionan no demuestran una regla**.

## 7. Una gráfica: ¿qué tan rápido crecen las potencias?

Contemos las cifras de 2ⁿ, 3ⁿ y 10ⁿ para n = 1, …, 50 y grafiquémoslas.

````python
import matplotlib.pyplot as plt

ns = list(range(1, 51))
cifras2 = [len(str(2 ** n)) for n in ns]
cifras3 = [len(str(3 ** n)) for n in ns]
cifras10 = [len(str(10 ** n)) for n in ns]

plt.plot(ns, cifras2, "o-", markersize=3, label="2^n")
plt.plot(ns, cifras3, "s-", markersize=3, label="3^n")
plt.plot(ns, cifras10, "^-", markersize=3, label="10^n")
plt.xlabel("n")
plt.ylabel("Cantidad de cifras")
plt.title("¿Cuántas cifras tiene una potencia?")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

✏️ Las tres gráficas parecen **rectas**. ¿Cuál es más inclinada? ¿Cuántas cifras gana 10ⁿ en cada paso? ¿Y 2ⁿ, aproximadamente?
