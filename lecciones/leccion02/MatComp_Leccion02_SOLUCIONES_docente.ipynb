### Ejercicio 1 · Suma de impares

```python
suma = 0
for n in range(1, 11):
    impar = 2 * n - 1
    suma = suma + impar
    print(f"n={n}: suma de los primeros {n} impares = {suma}   n² = {n**2}")
```

**Respuesta:** se obtienen los cuadrados perfectos: 1 + 3 + … + (2n − 1) = **n²**.

**Nota didáctica:** la prueba visual es muy poderosa: un cuadrado de n × n puntos se agranda a (n+1) × (n+1) agregando una "L" de 2n + 1 puntos. Pida que la dibujen en el cuaderno.

### Ejercicio 3 · Fibonacci pares

```python
fib = [1, 1]
while len(fib) < 30:
    fib.append(fib[-1] + fib[-2])

posiciones_pares = []
for i, f in enumerate(fib, start=1):
    marca = "PAR" if f % 2 == 0 else ""
    if f % 2 == 0:
        posiciones_pares.append(i)
    print(f"F{i} = {f} {marca}")

print("Posiciones de los pares:", posiciones_pares)
```

**Respuesta:** los pares aparecen en las posiciones 3, 6, 9, …: **uno de cada tres**.

**Explicación:** la paridad sigue el patrón impar, impar, par, impar, impar, par, … porque impar + impar = par, impar + par = impar y par + impar = impar; luego vuelven dos impares y el ciclo se repite.

### Ejercicio 4 · El campeón de Collatz

```python
mejor_n, mejor_pasos = 1, 0
max_n, max_valor = 1, 1
for n in range(1, 1001):
    camino = collatz(n)
    pasos = len(camino) - 1
    if pasos > mejor_pasos:
        mejor_n, mejor_pasos = n, pasos
    if max(camino) > max_valor:
        max_n, max_valor = n, max(camino)

print(f"Más pasos: n = {mejor_n} con {mejor_pasos} pasos")
print(f"Valor más alto: n = {max_n} llega hasta {max_valor}")
```

**Respuesta:** 871 tarda **178 pasos**. El que sube más alto es **703**, que alcanza **250 504** (el 27 llega a 9232 en 111 pasos).

**Nota didáctica:** muchos estudiantes suponen que "más pasos" implica "sube más alto". Este ejercicio muestra que son dos récords distintos. Para los más rápidos: ¿y del 1 al 10 000? (6171, con 261 pasos).

### Ejercicio 5 · La última cifra de Fibonacci

```python
# Últimas cifras de Fibonacci: basta trabajar con el residuo entre 10
cifras = [1, 1]
while True:
    cifras.append((cifras[-1] + cifras[-2]) % 10)
    if cifras[-2] == 1 and cifras[-1] == 1:   # el patrón vuelve a empezar
        break
periodo = len(cifras) - 2
print("Las cifras se repiten cada", periodo, "términos")

# Predicción para F_1000: la posición 1000 equivale a la posición 1000 % 60
r = 1000 % periodo
prediccion = cifras[r - 1] if r != 0 else cifras[periodo - 1]
print("1000 % 60 =", r, "→ predicción de la última cifra de F1000:", prediccion)

# Comprobación directa
a, b = 1, 1
for i in range(998):
    a, b = b, a + b
print("F1000 tiene", len(str(b)), "cifras y su última cifra es", b % 10)
```

**Respuesta:** el periodo es **60** (se llama *periodo de Pisano* módulo 10). Como 1000 = 16·60 + 40, la última cifra de F₁₀₀₀ es la de F₄₀ = 102 334 155, es decir **5**. F₁₀₀₀ tiene 209 cifras.

**Nota didáctica:** la idea clave es que el par (últimas dos cifras) determina todo lo que sigue; como solo hay 100 pares posibles, el patrón **tiene** que repetirse. Es un buen anticipo del principio del palomar.

### Reto · La regla 5n + 1

```python
def collatz5(n, max_pasos=200):
    camino = [n]
    while n != 1 and len(camino) <= max_pasos:
        n = n // 2 if n % 2 == 0 else 5 * n + 1
        camino.append(n)
    return camino

llegan = []
for n in range(1, 21):
    if collatz5(n)[-1] == 1:
        llegan.append(n)
print("Llegan a 1 con 5n+1:", llegan)

# Buscar el ciclo de 13: paramos cuando un valor se repite
n, vistos = 13, []
while n not in vistos:
    vistos.append(n)
    n = n // 2 if n % 2 == 0 else 5 * n + 1
inicio = vistos.index(n)
print("Ciclo desde 13:", vistos[inicio:] + [n])
```

**Respuesta:** del 1 al 20 solo llegan a 1: 1, 2, 3, 4, 6, 8, 12, 15, 16 y 19. El 13 cae en el ciclo **13 → 66 → 33 → 166 → 83 → 416 → 208 → 104 → 52 → 26 → 13**; otros números (como 7) parecen crecer sin control y nunca terminan dentro del límite de pasos.

**Conclusión para discutir:** una regla parecida se comporta de forma totalmente distinta. Que 3n + 1 "siempre" llegue a 1 no es obvio; por eso la conjetura de Collatz es tan difícil. Aproveche para insistir en la diferencia entre evidencia y demostración.

**Por qué es importante `max_pasos`:** sin un límite, la función con el 7 podría no terminar nunca. Es una buena ocasión para hablar de ciclos infinitos y de cómo detener una celda en Colab (botón ■).
