## 📚 El concepto: ¿qué es una sucesión?

Una **sucesión** es una lista ordenada de números que sigue una regla: a₁, a₂, a₃, …

Hay dos formas de describir la regla:

| Tipo | Idea | Ejemplo |
|---|---|---|
| **Recursiva** | Cada término se calcula a partir de los anteriores | "Empieza en 1 y suma 2 cada vez": 1, 3, 5, 7, … |
| **Fórmula (término general)** | Calculas el término n directamente | aₙ = 2n − 1 |

En Python guardamos sucesiones en **listas**: valores entre corchetes, separados por comas.

## 1. Listas: la caja donde guardamos sucesiones

````python
# Una lista con los primeros números impares
impares = [1, 3, 5, 7, 9]
print(impares)
print("Primer elemento:", impares[0])    # ¡Python cuenta desde 0!
print("Último elemento:", impares[-1])   # -1 significa "el último"
print("Cantidad de elementos:", len(impares))
print("Suma de todos:", sum(impares))
````

````python
# Construir una sucesión con un ciclo: empezamos con una lista vacía
# y le "pegamos" (append) un término en cada vuelta.
cuadrados = []
for n in range(1, 11):
    cuadrados.append(n**2)
print(cuadrados)
````

## 2. Números triangulares

Los **números triangulares** cuentan cuántos puntos se necesitan para formar un triángulo:

```
T₁=1    T₂=3     T₃=6       T₄=10
 ●       ●        ●          ●
        ● ●      ● ●        ● ●
                ● ● ●      ● ● ●
                          ● ● ● ●
```

**Regla recursiva:** Tₙ = Tₙ₋₁ + n (para el siguiente triángulo agregas una fila de n puntos).

````python
# Dibujamos los triángulos con texto
for n in range(1, 6):
    print(f"T{n}:")
    for fila in range(1, n + 1):
        print(" " * (n - fila) + "● " * fila)
    print()
````

````python
# Generamos los triangulares con la regla recursiva
triangulares = []
T = 0
for n in range(1, 16):
    T = T + n          # agregamos una fila de n puntos
    triangulares.append(T)
print(triangulares)
````

### ¿Hay una fórmula?

Cuenta la leyenda que el pequeño **Gauss**, con unos 9 años, sumó 1 + 2 + … + 100 en segundos: emparejó el primero con el último (1 + 100), el segundo con el penúltimo (2 + 99)… ¡y todas las parejas suman 101! Hay 50 parejas, así que la suma es 50 × 101 = 5050.

En general: **Tₙ = n(n + 1) / 2**. Comprobémoslo con la computadora.

````python
# Comparamos la regla recursiva con la fórmula de Gauss
for n in range(1, 16):
    formula = n * (n + 1) // 2
    print(n, triangulares[n - 1], formula, triangulares[n - 1] == formula)
````

**¿Qué pasa si…?** (anota tu conjetura antes de ejecutar)
- …sumas **dos triangulares consecutivos**: T₁+T₂, T₂+T₃, T₃+T₄…? ¿Qué números obtienes?

````python
# Sumamos triangulares vecinos
for i in range(len(triangulares) - 1):
    print(triangulares[i], "+", triangulares[i + 1], "=", triangulares[i] + triangulares[i + 1])
````

## 3. La sucesión de Fibonacci

En 1202, Leonardo de Pisa (Fibonacci) planteó un problema sobre conejos que da lugar a esta sucesión:

**1, 1, 2, 3, 5, 8, 13, …**

**Regla recursiva:** cada término es la suma de los **dos anteriores**: Fₙ = Fₙ₋₁ + Fₙ₋₂.

Aparece en la naturaleza: el número de espirales de un girasol o de una piña suele ser un número de Fibonacci.

````python
# Generamos los primeros 20 números de Fibonacci
fib = [1, 1]                     # los dos primeros términos
for i in range(18):
    siguiente = fib[-1] + fib[-2]  # suma de los dos últimos
    fib.append(siguiente)
print(fib)
````

Ahora una exploración: dividamos cada número de Fibonacci entre el anterior: 1/1, 2/1, 3/2, 5/3, 8/5, …
¿Qué crees que pasa con esos cocientes?

````python
import matplotlib.pyplot as plt

cocientes = []
for i in range(1, len(fib)):
    cocientes.append(fib[i] / fib[i - 1])

for c in cocientes[:12]:
    print(round(c, 6))

plt.plot(range(1, len(cocientes) + 1), cocientes, "o-")
plt.xlabel("n")
plt.ylabel("F(n+1) / F(n)")
plt.title("Cocientes de Fibonacci consecutivos")
plt.grid(True)
plt.show()
````

Los cocientes se acercan cada vez más a un número especial: el **número áureo**

$$\varphi = \frac{1 + \sqrt{5}}{2} \approx 1{,}6180339887…$$

````python
phi = (1 + 5**0.5) / 2
print("Número áureo:", phi)
print("Último cociente:", cocientes[-1])
print("Diferencia:", abs(phi - cocientes[-1]))
````

**¿Qué pasa si…?**
- …empiezas con otros dos números, por ejemplo `[2, 7]` o `[10, 3]`, y aplicas la misma regla? ¿Los cocientes siguen acercándose a φ? Cámbialo en la celda de abajo y experimenta.

````python
# Cambia los dos números iniciales y observa
sucesion = [2, 7]
for i in range(25):
    sucesion.append(sucesion[-1] + sucesion[-2])

print(sucesion[:10])
print("Cociente final:", sucesion[-1] / sucesion[-2])
````

## 4. El misterio de Collatz

Volvamos a la pregunta inicial. Primero vamos a escribir una **función**: un pedazo de código con nombre que podemos reutilizar. Se define con `def`.

También usaremos `while`, que repite **mientras** una condición sea verdadera (no sabemos de antemano cuántos pasos habrá).

````python
def collatz(n):
    """Devuelve la lista de números que visita n hasta llegar a 1."""
    camino = [n]
    while n != 1:            # repetir mientras no hayamos llegado a 1
        if n % 2 == 0:       # si es par...
            n = n // 2
        else:                # si es impar...
            n = 3 * n + 1
        camino.append(n)
    return camino

print(collatz(6))
print(collatz(7))
````

````python
# ¿Acertaste con el 7? Contemos los pasos (el número de flechas)
camino = collatz(7)
print("Pasos:", len(camino) - 1)
print("Número más alto alcanzado:", max(camino))
````

Ahora prueba con **27**. Predice: ¿cuántos pasos? ¿Qué tan alto subirá?

````python
camino27 = collatz(27)
print("Pasos:", len(camino27) - 1)
print("Máximo alcanzado:", max(camino27))

plt.figure(figsize=(10, 4))
plt.plot(camino27, ".-")
plt.title("El vuelo de Collatz empezando en 27")
plt.xlabel("Paso")
plt.ylabel("Valor")
plt.grid(True)
plt.show()
````

Por eso a estas sucesiones se les llama **"números granizo"**: suben y bajan como un granizo dentro de una nube antes de caer.

Exploremos muchos números a la vez: para cada n del 1 al 1000, ¿cuántos pasos tarda en llegar a 1?

````python
N = 1000
numeros = list(range(1, N + 1))
pasos = [len(collatz(n)) - 1 for n in numeros]   # una forma compacta de hacer un for

plt.figure(figsize=(10, 4))
plt.scatter(numeros, pasos, s=3)
plt.title("Pasos de Collatz para n = 1 … 1000")
plt.xlabel("n")
plt.ylabel("Pasos hasta llegar a 1")
plt.show()

print("Todos llegaron a 1 ✔")
````

**¿Qué pasa si…?**
- …miras los números **potencia de 2** (2, 4, 8, 16, …)? ¿Cuántos pasos tardan? ¿Por qué?
- …comparas `collatz(n)` con `collatz(2*n)`? ¿Qué relación hay entre sus pasos?
- …cambias `N = 1000` por `N = 10000`? ¿Aparecen nuevas "franjas" en la gráfica?

````python
# Espacio libre para explorar tus preguntas "¿qué pasa si…?"
for n in [2, 4, 8, 16, 32, 64]:
    print(n, "→", len(collatz(n)) - 1, "pasos")
````
