## 📚 El concepto: ¿qué es una función?

Una **función** es una regla que a cada número de entrada **x** le asigna **un único** número de salida **f(x)**. Puedes pensarla como una máquina:

> x → **[ f ]** → f(x)

Ejemplos:
- f(x) = 2x + 1: entra 3, sale 7.
- g(x) = x²: entra −4, sale 16.
- h(x) = 2ˣ: entra 10, sale 1024.

La **gráfica** de f es el conjunto de puntos (x, f(x)) en el plano. Nos deja **ver** cómo se comporta la función de un vistazo.

## 1. Funciones en Python y tablas de valores

```python
def f(x):
    return 2 * x + 1

def g(x):
    return x**2

print("x    f(x)   g(x)")
for x in range(-3, 4):
    print(f"{x:2}   {f(x):4}   {g(x):4}")
```

Para graficar necesitamos **muchos** valores de x. La librería `numpy` crea listas de números (llamadas *arreglos*) y hace cuentas con todos a la vez.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 7)      # 7 números igualmente espaciados entre -3 y 3
print(x)
print(2 * x + 1)               # ¡numpy calcula f para todos a la vez!
print(x**2)
```

```python
x = np.linspace(-3, 3, 200)

plt.figure(figsize=(7, 5))
plt.plot(x, f(x), label="f(x) = 2x + 1")
plt.plot(x, g(x), label="g(x) = x²")
plt.axhline(0, color="black", lw=0.8)
plt.axvline(0, color="black", lw=0.8)
plt.grid(True, alpha=0.4)
plt.legend()
plt.title("Una recta y una parábola")
plt.show()
```

## 2. La recta: y = mx + b

Mueve los sliders. **m** es la **pendiente** (cuánto sube y por cada paso de 1 en x) y **b** es el **intercepto** (dónde corta el eje y).

```python
from ipywidgets import interact, FloatSlider

def dibujar_recta(m, b):
    x = np.linspace(-5, 5, 100)
    plt.figure(figsize=(6, 6))
    plt.plot(x, m * x + b, lw=2.5, color="#2a6f97")
    # triangulito de pendiente: avanzar 1 en x, subir m en y
    plt.plot([1, 2, 2], [m + b, m + b, 2 * m + b], color="#e76f51", lw=2)
    plt.text(2.1, 1.5 * m + b, f"sube {m:.1f}", color="#e76f51")
    plt.scatter([0], [b], color="black", zorder=3)
    plt.axhline(0, color="black", lw=0.8); plt.axvline(0, color="black", lw=0.8)
    plt.xlim(-5, 5); plt.ylim(-10, 10)
    plt.grid(True, alpha=0.4)
    plt.title(f"y = {m:.1f}x {b:+.1f}")
    plt.show()

interact(dibujar_recta,
         m=FloatSlider(value=1, min=-4, max=4, step=0.1, continuous_update=False),
         b=FloatSlider(value=0, min=-6, max=6, step=0.5, continuous_update=False));
```

**¿Qué pasa si…?**
- …pones **m = 0**? ¿Y m negativo?
- …dejas m fijo y solo cambias b? ¿Qué tienen en común todas esas rectas?

**Un ejemplo real.** Un taxi cobra $4 500 al subir (banderazo) y $1 200 por kilómetro. El costo es una función lineal: **costo(km) = 1200 · km + 4500**. ¿Quién es m y quién es b?

```python
def costo_taxi(km):
    return 1200 * km + 4500

km = np.linspace(0, 20, 100)
plt.figure(figsize=(8, 4))
plt.plot(km, costo_taxi(km))
plt.xlabel("kilómetros"); plt.ylabel("costo ($)")
plt.title("Costo de un viaje en taxi")
plt.grid(True, alpha=0.4)
plt.show()

for k in [0, 5, 10, 15]:
    print(f"{k:2} km → ${costo_taxi(k):,}")
```

## 3. La parábola: y = ax² + bx + c

Las parábolas aparecen al lanzar una pelota, en las antenas parabólicas y en los puentes colgantes. Explora los tres parámetros. El punto rojo es el **vértice** (el punto más alto o más bajo), que está en x = −b / (2a). Los puntos negros son los **cortes con el eje x** (donde y = 0), si existen.

```python
def dibujar_parabola(a, b, c):
    x = np.linspace(-6, 6, 300)
    y = a * x**2 + b * x + c
    plt.figure(figsize=(6, 6))
    plt.plot(x, y, lw=2.5, color="#2a9d8f")
    if a != 0:
        xv = -b / (2 * a) + 0.0          # + 0.0 evita que se muestre -0.00
        yv = a * xv**2 + b * xv + c
        plt.scatter([xv], [yv], color="crimson", zorder=3, s=60)
        disc = b**2 - 4 * a * c            # el discriminante
        if disc >= 0:
            raices = [(-b - disc**0.5) / (2 * a), (-b + disc**0.5) / (2 * a)]
            plt.scatter(raices, [0, 0], color="black", zorder=3)
            texto = "cortes en x = " + ", ".join(f"{r:.2f}" for r in sorted(set(raices)))
        else:
            texto = "no corta el eje x"
        plt.title(f"y = {a:.1f}x² {b:+.1f}x {c:+.1f}\nvértice ({xv + 0:.2f}, {yv + 0:.2f}); {texto}")
    plt.axhline(0, color="black", lw=0.8); plt.axvline(0, color="black", lw=0.8)
    plt.xlim(-6, 6); plt.ylim(-10, 10)
    plt.grid(True, alpha=0.4)
    plt.show()

interact(dibujar_parabola,
         a=FloatSlider(value=1, min=-3, max=3, step=0.1, continuous_update=False),
         b=FloatSlider(value=0, min=-6, max=6, step=0.5, continuous_update=False),
         c=FloatSlider(value=-4, min=-8, max=8, step=0.5, continuous_update=False));
```

**¿Qué pasa si…?**
- …cambias el **signo** de a? ¿Y si a está muy cerca de 0?
- …mueves solo c? ¿Cuándo deja la parábola de cortar el eje x?
- …mueves solo b? ¿Qué camino sigue el vértice? (¡También es una parábola!)

```python
# Una pelota lanzada hacia arriba a 20 m/s: altura(t) = 20t − 4,9t²
t = np.linspace(0, 4.2, 200)
altura = 20 * t - 4.9 * t**2

plt.figure(figsize=(8, 4))
plt.plot(t, altura)
plt.axhline(0, color="black", lw=0.8)
plt.xlabel("tiempo (s)"); plt.ylabel("altura (m)")
plt.title("Altura de una pelota lanzada hacia arriba")
plt.grid(True, alpha=0.4)
plt.show()

t_max = 20 / (2 * 4.9)
print(f"Altura máxima: {20 * t_max - 4.9 * t_max**2:.2f} m a los {t_max:.2f} s")
print(f"Vuelve al suelo a los {20 / 4.9:.2f} s")
```

## 4. La carrera de crecimientos

Comparemos tres funciones que crecen: **lineal** (10x), **cuadrática** (x²) y **exponencial** (2ˣ). Mueve el slider para ver la carrera cada vez más lejos.

```python
from ipywidgets import IntSlider

def carrera(x_max):
    x = np.linspace(0, x_max, 400)
    plt.figure(figsize=(9, 5))
    plt.plot(x, 10 * x, label="lineal: 10x")
    plt.plot(x, x**2, label="cuadrática: x²")
    plt.plot(x, 2**x, label="exponencial: 2ˣ")
    plt.ylim(0, max(10 * x_max, x_max**2) * 1.5)
    plt.xlabel("x"); plt.legend(); plt.grid(True, alpha=0.4)
    plt.title(f"¿Quién va ganando entre 0 y {x_max}?")
    plt.show()

interact(carrera, x_max=IntSlider(value=5, min=2, max=20, continuous_update=False));
```

Al principio la lineal va adelante, luego la cuadrática… pero **la exponencial siempre termina ganando**, por muy grande que sea la ventaja inicial de las otras.

Volvamos a la oferta del millonario (en pesos, con 1 centavo = $0,01):

```python
dias = np.arange(1, 31)
opcion_a = 100_000 * dias                     # lo acumulado con la opción A
opcion_b = 0.01 * (2**dias - 1)                 # 0,01 + 0,02 + 0,04 + ... (¡Lección 5!)

plt.figure(figsize=(9, 4))
plt.plot(dias, opcion_a, "o-", label="A: $100 000 diarios")
plt.plot(dias, opcion_b, "o-", label="B: centavo que se duplica")
plt.xlabel("día"); plt.ylabel("total acumulado ($)")
plt.legend(); plt.grid(True, alpha=0.4)
plt.title("La oferta del millonario")
plt.show()

for d in [10, 20, 25, 30]:
    print(f"Día {d}: A = ${opcion_a[d - 1]:,.0f}   B = ${opcion_b[d - 1]:,.2f}")
```

## 5. Encontrar ceros: el método de bisección

A veces queremos saber **dónde una función vale 0**, pero no hay una fórmula sencilla. Por ejemplo, ¿para qué x se cumple x² = 2? Es decir, ¿dónde corta el eje x la función f(x) = x² − 2?

**Idea:** f(1) = −1 es negativo y f(2) = 2 es positivo, así que el cero está **entre 1 y 2**. Miramos el punto medio y nos quedamos con la mitad donde la función cambia de signo. Repetimos.

```python
def biseccion(f, a, b, pasos=30, mostrar=False):
    """Busca un cero de f entre a y b (f(a) y f(b) deben tener signos distintos)."""
    for i in range(pasos):
        m = (a + b) / 2
        if mostrar:
            print(f"paso {i + 1:2}: el cero está entre {a:.6f} y {b:.6f}")
        if f(a) * f(m) <= 0:      # cambia de signo en la mitad izquierda
            b = m
        else:                     # cambia de signo en la mitad derecha
            a = m
    return (a + b) / 2

def h(x):
    return x**2 - 2

raiz = biseccion(h, 1, 2, pasos=10, mostrar=True)
print("Aproximación:", raiz)
print("Raíz de 2 con 30 pasos:", biseccion(h, 1, 2))
print("Python dice:          ", 2**0.5)
```

Cada paso **parte el intervalo por la mitad**: después de 10 pasos el intervalo mide 1/2¹⁰ ≈ 0,001, y después de 30, menos de una milmillonésima.
